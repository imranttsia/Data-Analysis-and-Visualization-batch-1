#Data Visualization

Data visualization is the process of translating data into a visual context, such as a map, graph, or chart. It helps people understand the significance of data by placing it in a visual context. This makes complex data more accessible, understandable, and usable. Visualizations reveal trends, patterns, and outliers that might otherwise be missed in text-based data. It's a crucial step in data analysis, allowing for quicker insights and better decision-making across various fields.

#Matplotlib

Matplotlib is a comprehensive library for creating static, animated, and interactive visualizations in Python. It is a powerful and popular tool for data visualization, offering a wide array of plotting functions to generate various types of plots like line plots, scatter plots, bar charts, histograms, and more. It is highly customizable, allowing users fine-grained control over every aspect of a plot.

### Use Cases of Data Visualization and Matplotlib

Data visualization and Matplotlib are widely used across various domains due to their ability to transform complex data into understandable visual insights. Here are some key use cases:

    *   **Medical Imaging:** Representing medical scan data (though specialized tools are often used, Matplotlib can be a foundational step for custom visualizations).

3.  **Data Science and Machine Learning:**
    *   **Exploratory Data Analysis (EDA):** Understanding data distributions (histograms), relationships between variables (scatter plots), and identifying outliers before model building.
    *   **Model Evaluation:** Visualizing model performance metrics (e.g., ROC curves, precision-recall curves), error distributions, and prediction accuracy.
    *   **Feature Engineering:** Understanding the impact of different features on the target variable.

4.  **Healthcare and Medicine:**
    *   **Patient Data Analysis:** Tracking patient vital signs, disease progression, and treatment effectiveness over time.

5.  **Education and Journalism:**
    *   **Explaining Concepts:** Creating clear and intuitive graphs to illustrate complex theories or data in textbooks, presentations, or news articles.
    *   **Data-Driven Storytelling:** Journalists use visualizations to present data-backed stories, making information more engaging and accessible to the public.

6.  **Environmental Science:**
    *   **Climate Change Analysis:** Plotting temperature anomalies, sea-level rise, and carbon emissions over decades.
    *   **Pollution Monitoring:** Visualizing pollution levels in air or water across different locations or times.

These examples highlight how Matplotlib's flexibility allows users to create a wide array of static, animated, and interactive visualizations tailored to specific needs, making it an indispensable tool for data analysis and communication.

#matplotlib and NumPy

Even the Matplot library, despite being a fully graphical library, has its foundation the NumPy library. In fact, you have seen so far how to pass lists as arguments, both to represent the data and to set the extremes of the axes. Actually, these lists have been converted internally in NumPy arrays.

Therefore, you can directly enter NumPy arrays as input data. This array of data, which have been processed by pandas, can be directly used with matplotlib without further processing.

As an example now you see how it is possible to plot three different trends in the same plot. You can choose for this example the sin() function belonging to the math module. So you will need to import it. To generate points following a sinusoidal trend you will use the library NumPy.

Generate a series of points on the x axis using the arange() function, while for the values on the y axis you will use the map() function to apply the sin() function on all the items of the array (without using a for loop).

```python
import math
import numpy as np
t = np.arange(0,2.5,0.1)
y1 = list(map(math.sin,math.pi*t))
y2 = list(map(math.sin,math.pi*t+math.pi/2))
y3 = list(map(math.sin,math.pi*t-math.pi/2))
plt.plot(t,y1,'b*',t,y2,'g^',t,y3,'ys')
```

This code generates three different sinusoidal trends on the same plot using `numpy` and `matplotlib.pyplot`.

*   **`import math` and `import numpy as np`**: Imports the necessary libraries for mathematical functions and numerical array operations.
*   **`t = np.arange(0, 2.5, 0.1)`**: Creates a NumPy array `t` which serves as the x-axis values, ranging from 0 to 2.4 with a step of 0.1.
*   **`y1 = list(map(math.sin, math.pi * t))`**: Calculates the sine values for `y1`. It multiplies `t` by `math.pi` and then applies the `math.sin` function to each element using `map()`. The result is converted to a list.
*   **`y2 = list(map(math.sin, math.pi * t + math.pi / 2))`**: Similar to `y1`, but `math.pi / 2` is added to the argument, creating a phase shift.
*   **`y3 = list(map(math.sin, math.pi * t - math.pi / 2))`**: Similar to `y1` and `y2`, but `math.pi / 2` is subtracted, creating another phase shift.
*   **`plt.plot(t, y1, 'b*', t, y2, 'g^', t, y3, 'ys')`**: This command plots all three trends on the same graph:
    *   `(t, y1, 'b*')`: Plots `t` against `y1` using blue asterisks.
    *   `(t, y2, 'g^')`: Plots `t` against `y2` using green upward-pointing triangles.
    *   `(t, y3, 'ys')`: Plots `t` against `y3` using yellow squares.

As you can see in the plot represents the three different temporal trends with three different colors and markers. In these cases, when the trend of a function is so obvious, the plot is perhaps not the most appropriate representation, but it is better to use the lines. To differentiate the three trends as well as using different colors, you can use the pattern composed of different combinations of dots
and dashes ( - and . ).

```python
plt.plot(t,y1,'b--',t,y2,'g',t,y3,'r-.')
```

#Working with Multiple Figures and Axes

So far you have seen how all pyplot commands are routed to the display of a single figure. Actually, matplotlib allows management of multiple figures simultaneously, and within each figure, it offers the ability to view different plots defined as subplots.

So when you are working with pyplot, you must always keep in mind the concept of current Figure and current Axes (that is, the plot shown within the figure).

Now you will see an example where two subplots are represented in a single figure. The subplot() function, in addition to subdividing the figure in different drawing areas, is used at the same time in order to focus the commands on a specific subplot.

The argument passed to the subplot() function sets the mode of subdivision and which is the current subplot. The current subplot will be the only figure which will be affected by the commands. The argument
of the subplot() function is composed of three integers. The first number defines how many parts the figure is split into vertically. The second number defines how many parts the figure is divided into horizontally.

The third issue selects which is the current subplot on which you can direct commands.

Now you will display two sinusoidal trends (sine and cosine) and the best way to do that is to divide the canvas vertically in two horizontal subplots (as shown in Figure 7-12). So the numbers to pass as an
argument are ‘211’ and ‘212’ .

```python
import numpy as np
t = np.arange(0,5,0.1)
y1 = np.sin(2*np.pi*t)
y2 = np.sin(2*np.pi*t)
```

```python
plt.subplot(211)
plt.plot(t,y1,'b-.')
plt.subplot(212)
plt.plot(t,y2,'r--')
```

Now you do the same thing by dividing the figure in two vertical subplots. The numbers to be passed as arguments to the subplot() function are ‘121’ and ‘122’

```python
t = np.arange(0.,1.,0.05)
y1 = np.sin(2*np.pi*t)
y2 = np.cos(2*np.pi*t)
```

```python
plt.subplot(121)
plt.plot(t,y1,'b-.')
plt.subplot(122)
plt.plot(t,y2,'r--')
```

#Adding Further Elements to the Chart

In order to make a chart more informative, many times it is not enough to represent the data using lines or markers and assign the range of values using two axes. In fact, there are many other elements that can be added to a chart in order to enrich it with additional information.

In this section you will see how to add additional elements to the chart as text labels, a legend, and so on.

#Adding Text

You’ve already seen how you can add the title to a chart with the title() function. Two other textual indications very important to be added to a chart are the axis labels. This is possible through the use of two other specific functions such as xlabel() and ylabel(). These functions take as argument a string which will be the shown text.

Now add two axis labels to the chart. They will describe which kind of value is assigned to each axis

```python
plt.axis([0,5,0,20])
plt.title('My first plot')
plt.xlabel('Counting')
plt.ylabel('Square values')
plt.plot([1,2,3,4],[1,4,9,16],'ro')
```

Thanks to the keywords, you can change the characteristics of the text. For example, you can modify the title by changing the font and increasing the size of the characters. Also you can modify the color of the axis labels to bring out more the title of the plot

```python
plt.axis([0,5,0,20])
plt.title('My first plot',fontsize=20,fontname='Times New Roman')
plt.xlabel('Counting',color='gray')
plt.ylabel('Square values',color='gray')
plt.plot([1,2,3,4],[1,4,9,16],'ro')
```

But matplotlib is not limited to this: pyplot allows you to add text to any position within a chart.

This feature is performed by a specific function called text().

text(x,y,s, fontdict=None, **kwargs)

The first two arguments are the coordinates of the location where you want to place the text. s is the string of text to be added, and fontdict (optional) is the font that you want the text to be represented. Finally, you can add the keywords.

Thus, add the label to each point of the plot. Because the first two arguments to the text() function are the coordinates of the graph, you have to use the coordinates of the four points of the plot shifted slightly on the y axis.

```python
plt.axis([0,5,0,20])
plt.title('My first plot',fontsize=20,fontname='Times New Roman')
plt.xlabel('Counting',color='gray')
plt.ylabel('Square values',color='gray')
plt.text(1,1.5,'First')
plt.text(2,4.5,'Second')
plt.text(3,9.5,'Third')
plt.text(4,16.5,'Fourth')
plt.plot([1,2,3,4],[1,4,9,16],'ro')
```

Since matplotlib is a graphics library designed to be used in scientific circles, it must be able to exploit the full potential of scientific language, including mathematical expressions. Matplotlib offers the possibility to integrate LaTeX expressions allowing you to insert mathematical expressions within the chart.

To do this you can add a LaTeX expression to the text, enclosing it between two ‘$’ characters. The interpreter will recognize them as LaTeX expressions and convert them in the corresponding graphic, which can be a mathematical expression, a formula mathematical characters, or just Greek letters. Generally you have to precede the string containing LaTeX expressions with an ‘r’, which indicates raw text in order to avoid unintended escape sequences.

Here, you can also make use of the keywords to further enrich the text to be shown in the plot.

Therefore, as an example, you can add the formula describing the trend followed by the point of the plot and enclose it in a colored bounding box

```python
plt.axis([0,5,0,20])
plt.title('My first plot',fontsize=20,fontname='Times New Roman')
plt.xlabel('Counting',color='gray')
plt.ylabel('Square values',color='gray')
plt.text(1,1.5,'First')
plt.text(2,4.5,'Second')
plt.text(3,9.5,'Third')
plt.text(4,16.5,'Fourth')
plt.text(1.1,12,r'$y = x^2$',fontsize=20,bbox={'facecolor':'yellow','alpha':0.2})
plt.plot([1,2,3,4],[1,4,9,16],'ro')
```

In the plt.text() function, when you use bbox={'facecolor':'yellow','alpha':0.2}, the alpha parameter controls the transparency of the bounding box's background color. The value ranges from 0.0 (completely transparent) to 1.0 (completely opaque). In your example, alpha:0.2 means the yellow background of the text box will be 20% opaque (or 80% transparent), allowing some of the underlying plot to show through.

#Adding a Grid

Another element you can add to a plot is a grid. Often its addition is necessary in order to better understand
the position occupied by each point on the chart.

Adding a grid to a chart is a very simple operation: just add the grid() function, passing True as argument

```python
plt.axis([0,5,0,20])
plt.title('My first plot',fontsize=20,fontname='Times New Roman')
plt.xlabel('Counting',color='gray')
plt.ylabel('Square values',color='gray')
plt.text(1,1.5,'First')
plt.text(2,4.5,'Second')
plt.text(3,9.5,'Third')
plt.text(4,16.5,'Fourth')
plt.text(1.1,12,r'$y = x^2$',fontsize=20,bbox={'facecolor':'yellow','alpha':0.2})
plt.grid(True)
plt.plot([1,2,3,4],[1,4,9,16],'ro')
```

#Adding a Legend

Another very important component that should be present in any chart is the legend. pyplot also provides a specific function for this type of object: legend().

Add a legend to your chart with the legend() function and a string indicating the words with which you want the series to be shown. In this example you will assign the 'First series' name to the input data array

```python
plt.axis([0,5,0,20])
plt.title('My first plot',fontsize=20,fontname='Times New Roman')
plt.xlabel('Counting',color='gray')
plt.ylabel('Square values',color='gray')
plt.text(2,4.5,'Second')
plt.text(3,9.5,'Third')
plt.text(4,16.5,'Fourth')
plt.text(1.1,12,'$y = x^2$',fontsize=20,bbox={'facecolor':'yellow','alpha':0.2})
plt.grid(True)
plt.plot([1,2,3,4],[1,4,9,16],'ro')
plt.legend(['First series'])
```

For example, the position occupied by the legend is set by assigning numbers from 0 to 10 to the loc kwarg. Each of these numbers characterizes one of the corners of the chart. A value of 1 is the default, that is, the upper-right corner. In the next
example you will move the legend in the upper-left corner so it will not overlap with the points represented in the plot.

Handling Date Values

One of the most common problems encountered when doing data analysis is handling data of the date-time type. Displaying them along an axis (normally the x axis) can be really problematic especially for the management of ticks.
Take for example the display of a linear chart with a data set of eight points in which you have to represent date values on the x axis with the following format: day-month-year.

```python
import datetime
import numpy as np
import matplotlib.pyplot as plt
events = [datetime.date(2015,1,23),datetime.date(2015,1,28),datetime.date(2015,2,3),datetime.date(2015,2,21),datetime.date(2015,3,15),datetime.date(2015,3,24),datetime.date(2015,4,8),datetime.date(2015,4,24)]
readings = [12,22,25,20,18,15,17,14]
plt.plot(events,readings)
```

As you can see in Figure, the automatic management of ticks and especially the tick labels, it is a disaster. The dates expressed in this way are difficult to read, there are no clear time intervals elapsed between one point and another, and there is also overlap.

To manage the dates therefore it is advisable to define a time scale with appropriate objects. First you need to import matplotlib.dates, a module specialized for the management of this type of data. Then you define the scales of the times, as in this case, a scale of days and one of the months, through the functions
MonthLocator() and DayLocator(). In these cases, the formatting is also very important, and to avoid overlap or unnecessary references, you have to limit the tick labels to the essential, which in this case is year-month. This format can be passed as an argument to DateFormatter() function.

After you defined the two scales, one for the days and one for the months, you can set two different kinds of ticks on the x axis, using the functions set_major_locator() and set_minor_locator() on the xaxis
object. Instead, to set the text format of the tick labels referred to the months you have to use the set_major_formatter() function.

Making all these settings you finally obtain the plot as shown in Figure.

```python
import datetime
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
months = mdates.MonthLocator()
days = mdates.DayLocator()
timeFmt = mdates.DateFormatter('%Y-%m')
events = [datetime.date(2015,1,23),datetime.date(2015,1,28),datetime.date(2015,2,3),datetime.date(2015,2,21),datetime.date(2015,3,15),datetime.date(2015,3,24),datetime.date(2015,4,8),datetime.date(2015,4,24)]
readings = [12,22,25,20,18,15,17,14]
fig, ax = plt.subplots()
plt.plot(events,readings)
ax.xaxis.set_major_locator(months)
ax.xaxis.set_major_formatter(timeFmt)
ax.xaxis.set_minor_locator(days)
```

#Line Chart

Among all the types of chart the linear chart is the simplest. A line chart is a sequence of data points connected by a line. Each data point consists of a pair of values (x,y), which will be reported in the chart according to the scale of values of the two axes (x and y).

By way of example you can begin to plot the points generated by a mathematical function. Then, you can consider a generic mathematical function such as

y = sin (3 * x) / x

Therefore, if you want to create a sequence of data points, you need to create two NumPy arrays. First you create an array containing the x values to be referred to the x axis. In order to define a sequence of increasing values you will use the np.arange() function. Since the function is sinusoidal you should refer to values that are multiples and submultiples of the greek pi (np.pi). Then using these sequence of values you can obtain the y values applying the np.sin() function directly to these values (Thanks to NumPy!).

After all this, you have only to plot them calling the plot() function. You will obtain a line chart as shown in Figure

```python
import matplotlib.pyplot as plt
import numpy as np
x = np.arange(-2*np.pi,2*np.pi,0.01)
y = np.sin(3*x)/x
plt.plot(x,y)
```

Now you can extend the case in which you want to display a family of functions such as

In all the linear charts you have seen so far, you always have the x axis and y axis placed at the edge of the Figure (corresponding to the sides of the bounding border box). Another way of displaying axes is to have the two axes passing through the origin (0, 0), i.e., the two Cartesian axes.

To do this, you must first capture the Axes object through the gca() function. Then through this object, you can select each of the four sides making up the bounding box, specifying for each one its position: right,
left, bottom, and top. Crop the sides that do not match any axis (right and bottom) using the set_color() function, indicating none as color. Then, the sides corresponding to the x and y axes are moved to pass through the origin (0,0) with the set_position() function.

```python
import matplotlib.pyplot as plt
import numpy as np
x = np.arange(-2*np.pi,2*np.pi,0.01)
y = np.sin(3*x)/x
y2 = np.sin(2*x)/x
y3 = np.sin(4*x)/x
plt.plot(x,y,color='b')
plt.plot(x,y2,color='r')
plt.plot(x,y3,color='g')
plt.xticks([-2*np.pi, -np.pi, 0, np.pi, 2*np.pi],[r'$-2\pi$',r'$-\pi$',r'$0$',r'$+\pi$',r'$+2\pi$'])
plt.yticks([-1,0,+1,+2,+3],[r'$-1$',r'$0$',r'$+1$',r'$+2$',r'$+3$'])
ax = plt.gca()
ax.spines['right'].set_color('none')
ax.spines['top'].set_color('none')
ax.xaxis.set_ticks_position('bottom')
ax.spines['bottom'].set_position(('data',0))
ax.yaxis.set_ticks_position('left')
ax.spines['left'].set_position(('data',0))
```

This code builds upon the previous contour plot example by customising the appearance and position of the x and y axes to mimic standard Cartesian axes passing through the origin (0,0).

*   **`ax.spines['right'].set_color('none')`** and **`ax.spines['top'].set_color('none')`**: These lines make the 'right' and 'top' spines (lines enclosing the plot area) invisible by setting their color to 'none'. This creates the appearance of only having two axes.
*   **`ax.xaxis.set_ticks_position('bottom')`**: This command places the tick marks for the x-axis on the 'bottom' spine.
*   **`ax.spines['bottom'].set_position(('data',0))`**: This moves the 'bottom' spine (the x-axis) to intersect the y-axis at `y=0` in data coordinates.
*   **`ax.yaxis.set_ticks_position('left')`**: This command places the tick marks for the y-axis on the 'left' spine.
*   **`ax.spines['left'].set_position(('data',0))`**: This moves the 'left' spine (the y-axis) to intersect the x-axis at `x=0` in data coordinates.

In summary, this code transforms the plot from having axes on the bounding box edges to having them intersect at the origin (0,0), a common layout for mathematical function plots.

Often, it is very useful to be able to specify a particular point of the line using a notation and optionally adding an arrow to better indicate the position of the point. In this regard matplotlib provides a function called annotate(), especially useful in these cases, even if the numerous kwargs needed to obtain a good result can make its settings quite complex. The first argument is the string to be represented containing the expression in LaTeX; then you can add the various kwargs. The point of the chart to note is indicated by a list containing the coordinates of the point [x, y] passed to the xy kwarg. The distance of the textual notation from the point to be highlighted is defined by the xytext kwarg
and represented by means of a curved arrow whose characteristics are defined in the arrowprops kwarg.

```python
import matplotlib.pyplot as plt
import numpy as np
x = np.arange(-2*np.pi,2*np.pi,0.01)
y = np.sin(3*x)/x
y2 = np.sin(2*x)/x
y3 = np.sin(4*x)/x
plt.plot(x,y,color='b')
plt.plot(x,y2,color='r')
plt.plot(x,y3,color='g')
plt.xticks([-2*np.pi, -np.pi, 0, np.pi, 2*np.pi],[r'$-2\pi$',r'$-\pi$',r'$0$',r'$+\pi$',r'$+2\pi$'])
plt.yticks([-1,0,+1,+2,+3],[r'$-1$',r'$0$',r'$+1$',r'$+2$',r'$+3$'])
plt.annotate(r'$\lim_{x\to 0}\frac{\sin(x)}{x}= 1$', xy=[0,1],xycoords='data', xytext=[30,30],fontsize=16,textcoords='offset points',arrowprops=dict(arrowstyle="->",connectionstyle="arc3,rad=.2"))
ax = plt.gca()
ax.spines['right'].set_color('none')
ax.spines['top'].set_color('none')
ax.xaxis.set_ticks_position('bottom')
ax.spines['bottom'].set_position(('data',0))
ax.yaxis.set_ticks_position('left')
ax.spines['left'].set_position(('data',0))
```

#Histogram

A histogram consists of adjacent rectangles erected on the x axis, split into discrete intervals called bins, and with an area proportional to the frequency of the occurrences for that bin. This kind of visualization is commonly used in statistical studies about distribution of samples.

In order to represent a histogram pyplot provides a special function called hist(). This graphic function also has a feature that other functions producing charts do not have. The hist() function, in addition to drawing the histogram, returns a tuple of values that are the results of the calculation of the histogram. In fact the hist() function can also implement the calculation of the histogram, that is, it is sufficient to provide a series of samples of values as an argument and the number of bins in which to be divided, and it will take care of dividing the range of samples in many intervals (bins), and then calculate the occurrences for each bin. The result of this operation, in addition to being shown in graphical form (see Figure), will be returned in the form of a tuple.

(n, bins, patches)

To see this operation there is no better explanation of a practical example. Then generate a population of 100 random values from 0 to 100 using the random.randint() function.

```python
import matplotlib.pyplot as plt
import numpy as np
pop = np.random.randint(0,100,100)
pop
```

Now, create the histogram of these samples by passing as an argument of the hist(). function. For example, you want to divide the occurrences in 20 bins (if not specified, the default value is 10 bins) and to do that you have to use the kwarg bin

```python
n,bins,patches = plt.hist(pop,bins=20)
```

This line creates a histogram and unpacks its results into three variables: `n`, `bins`, and `patches`.

*   **`plt.hist(pop, bins=20)`**: This is the core function call:
    *   `pop`: The input data, a NumPy array of 100 random integers between 0 and 100.
    *   `bins=20`: Specifies that the data should be divided into 20 equal-width intervals (or 'bins'). Matplotlib automatically determines the range of these bins based on your `pop` data.

*   **`n`**: This variable receives the counts of data points within each bin. It's a NumPy array where each value represents how many data points from `pop` fell into the corresponding bin.
    *   *From the kernel state, `n` is `array([ 4.,  6.,  5.,  4.,  4.,  5.,  6., 11.,  7.,  4.,  3.,  2.,  3.,  6.,  6.,  1.,  7.,  3.,  8.,  5.])`. This means, for example, 4 data points fell into the first bin, 6 into the second, and so on.*

*   **`bins`**: This variable receives the edges of the bins. It's a NumPy array that defines the boundaries of each bin. Since there are 20 bins, there will be 21 bin edges (start of the first bin to the end of the last bin).
    *   *From the kernel state, `bins` is `array([ 1.,  5.9, 10.8, 15.7, 20.6, 25.5, 30.4, 35.3, 40.2, 45.1, 50., 54.9, 59.8, 64.7, 69.6, 74.5, 79.4, 84.3, 89.2, 94.1, 99.])`. This indicates the range for each of the 20 bins.*

In summary, this line calculates the frequency distribution of your `pop` data across 20 bins, visualizes it as a histogram, and provides the raw counts, bin boundaries, and graphical elements for further analysis or customization.

Actually there are many other steps we can take to further refine the bar chart. Each of these finishes is set by adding a specific kwarg as an argument in the bar() function. For example, you can add the standard deviation values of the bar through the yerr kwarg along with a list containing the standard deviations.     *   `index`: Passed as the first argument, these are the x-coordinates where the bars will be placed.
    *   `values1`: Passed as the second argument, these are the heights of the bars on the y-axis.
    *   `yerr=std1`: This keyword argument takes the `std1` list, which provides the error values for each bar, resulting in vertical error bars.

*   **`plt.xticks(index,['A','B','C','D','E'])`**: This customizes the x-axis tick marks:
    *   `index`: Specifies the positions where the tick marks should appear (`[0, 1, 2, 3, 4]`).
    *   `['A','B','C','D','E']`: These are the labels that will be displayed at each of the specified tick positions, effectively naming the categories for each bar.
*   **`plt.legend(loc=2)`**: This adds a legend to the plot.
    *   `loc=2`: Specifies the location of the legend. As discussed previously, `loc=2` places the legend in the upper-left corner of the plot.

#Multiserial Bar Chart

As line charts, bar charts also generally are used to simultaneously to display more series of values. But in this case it is necessary to make some clarifications on how to structure a multiseries bar chart. So far you have defined a sequence of indexes, each corresponding to a bar, to be assigned to the x axis. These indices should represent categories. In this case, however, we have more bars that must share the same category.

One approach used to overcome this problem is to divide the space occupied by an index (for convenience its width is 1) in as many parts as are the bars sharing that index and that we want to display. Moreover, it is advisable to add an additional space which will serve as gap to separate a category with respect to the next (as shown in Figure).

```python
import matplotlib.pyplot as plt
import numpy as np
index = np.arange(5)
values1 = [5,7,3,4,6]
values2 = [6,6,4,5,7]
values3 = [5,6,5,4,6]
bw = 0.3
plt.axis([-0.4,5,0,8])
plt.title('A Multiseries Bar Chart',fontsize=20)
plt.bar(index,values1,bw,color='b')
plt.bar(index+bw,values2,bw,color='g')
plt.bar(index+2*bw,values3,bw,color='r')
plt.xticks(index+bw,['A','B','C','D','E'])
```

This code creates a multiseries bar chart, where three different sets of values are displayed for the same categories. Here's a breakdown of what each part does:

*   **`import matplotlib.pyplot as plt`** and **`import numpy as np`**: Imports the necessary libraries for plotting and numerical operations.
*   **`index = np.arange(5)`**: Creates a NumPy array `[0, 1, 2, 3, 4]` to serve as the base positions for the groups of bars on the x-axis.
*   **`values1 = [5, 7, 3, 4, 6]`, `values2 = [6, 6, 4, 5, 7]`, `values3 = [5, 6, 5, 4, 6]`**: Defines three lists, each representing a series of values for the bars.
*   **`bw = 0.3`**: Sets the width of each individual bar.
*   **`plt.axis([0,5,0,8])`**: Sets the x and y axis limits to ensure all bars are visible with appropriate padding.
*   **`plt.title('A Multiseries Bar Chart',fontsize=20)`**: Sets the plot title and its font size.
*   **`plt.bar(index,values1,bw,color='b')`**: Plots the first series of bars (blue) at `index` positions.
*   **`plt.bar(index+bw,values2,bw,color='g')`**: Plots the second series of bars (green), shifted by `bw` to be adjacent to the first series.
*   **`plt.bar(index+2*bw,values3,bw,color='r')`**: Plots the third series of bars (red), shifted by `2*bw` to be adjacent to the second series.
*   **`plt.xticks(index+1.5*bw,['A','B','C','D','E'])`**: Customizes the x-axis tick labels, centering them within each group of three bars.

In summary, this code generates a grouped bar chart where for each category (A, B, C, D, E), three different values (series1, series2, series3) are displayed side-by-side, making it easy to compare the series within each category.

Regarding the multiseries horizontal bar chart (Figure), things are very similar. You have to replace the bar() function with the corresponding barh() function and also remember to replace the xticks() function with the yticks() function. You need to reverse the range of values covered by the axes in the axis() function.

```python
import matplotlib.pyplot as plt
import numpy as np
index = np.arange(5)
values1 = [5,7,3,4,6]
values2 = [6,6,4,5,7]
values3 = [5,6,5,4,6]
bw = 0.3
plt.axis([0,8,-0.4,5])
plt.title('A Multiseries Horizontal Bar Chart',fontsize=20)
plt.barh(index,values1,bw,color='b')
plt.barh(index+bw,values2,bw,color='g')
plt.barh(index+2*bw,values3,bw,color='r')
plt.yticks(index+bw,['A','B','C','D','E'])
```

Another form to represent a multiseries bar chart is in the stacked form, in which the bars are stacked one on the other. This is especially useful when you want to show the total value obtained by the sum of all the bars.

To transform a simple multiseries bar chart in a stacked one, you add the bottom kwarg to each bar() function. Each series must be assigned to the corresponding bottom kwarg. At the end you will obtain the stacked bar chart as shown in Figure

```python
import matplotlib.pyplot as plt
import numpy as np
series1 = np.array([3,4,5,3])
series2 = np.array([1,2,2,5])
series3 = np.array([2,3,3,4])
index = np.arange(4)
plt.axis([-0.5,3.5,0,15])
plt.bar(index,series1,color='r')
plt.bar(index,series2,color='b',bottom=series1)
plt.bar(index,series3,color='g',bottom=(series2+series1))
plt.xticks(index,['Jan15','Feb15','Mar15','Apr15'])
```

This code snippet uses Matplotlib and NumPy to generate and display a stacked bar chart. Here's a breakdown of what each part does:

*   **`import matplotlib.pyplot as plt`** and **`import numpy as np`**: These lines import the necessary libraries. `matplotlib.pyplot` is used for creating plots, and `numpy` is used for numerical operations, especially for creating arrays.
*   **`series1 = np.array([3,4,5,3])`**, **`series2 = np.array([1,2,2,5])`**, **`series3 = np.array([2,3,3,4])`**: These lines define three NumPy arrays. Each array represents a 'series' of data, where each number corresponds to a segment of a stacked bar.
*   **`index = np.arange(4)`**: This creates a NumPy array `[0, 1, 2, 3]`. This `index` array will be used to position the bars along the x-axis, representing four distinct categories (e.g., months).
*   **`plt.axis([-0.5,3.5,0,15])`**: This sets the limits for the x and y axes of the plot. `[-0.5, 3.5]` defines the x-axis range, ensuring all four bars are fully visible. `[0, 15]` defines the y-axis range.
*   **`plt.bar(index,series1,color='r')`**: This is the first **`plt.bar`** call, which creates the bottom layer of the stacked bars. It plots `series1` (the heights) at the `index` positions (the x-coordinates) with a red color (`'r'`).
*   **`plt.bar(index,series2,color='b',bottom=series1)`**: This creates the second layer of the stacked bars. It plots `series2` on top of `series1`. The `bottom=series1` argument is crucial here; it tells Matplotlib to start drawing these bars from the top of the `series1` bars, effectively stacking them. These bars are colored blue (`'b'`).
*   **`plt.bar(index,series3,color='g',bottom=(series2+series1))`**: This creates the third and final layer. It plots `series3` on top of the sum of `series1` and `series2`. The `bottom=(series2+series1)` argument ensures these bars start where the previous stack ended. These bars are colored green (`'g'`).
*   **`plt.xticks(index,['Jan15','Feb15','Mar15','Apr15'])`**: This customizes the x-axis tick marks. `index` specifies the positions (`0, 1, 2, 3`) where the tick marks should appear, and the list `['Jan15','Feb15','Mar15','Apr15']` provides the labels for those positions. This works well for a stacked bar chart where all parts of a stack share the same x-position.

Here too, in order to create the equivalent horizontal stacked bar chart, you need to replace the bar() function with barh() function, being careful to change other parameters as well. Indeed xticks() function
should be replaced with the yticks() function because the labels of the categories now must be reported on the
y axis. After doing all these changes you will obtain the horizontal stacked bar chart as shown in Figure

```python
import matplotlib.pyplot as plt
import numpy as np
index = np.arange(4)
series1 = np.array([3,4,5,3])
series2 = np.array([1,2,2,5])
series3 = np.array([2,3,3,4])
plt.axis([0,15,-0.5,4])
plt.title('A Multiseries Horizontal Stacked Bar Chart')
plt.barh(index,series1,color='r')
plt.barh(index,series2,color='g',left=series1)
plt.barh(index,series3,color='b',left=(series1+series2))
plt.yticks(index,['Jan15','Feb15','Mar15','Apr15'])
```

So far the various series have been distinguished by using different colors. Another mode of distinction between the various series is for example, using hatches that allow to fill the various bars with strokes drawn in a different way. To do this, you have first to set the color of the bar as white and then you have to use the hatch kwarg to define how the hatch is to be set. The various hatches have codes distinguishable among these characters (|, /, -, \, *, -) corresponding to the line style filling the bar. The more the same symbol is replicated, the denser will be the lines forming the hatch. For example, /// is more dense than //, which is
more dense than /. (See Figure)

```python
import matplotlib.pyplot as plt
import numpy as np
index = np.arange(4)
series1 = np.array([3,4,5,3])
series2 = np.array([1,2,2,5])
series3 = np.array([2,3,3,4])
plt.axis([0,15,-0.5,4])
plt.title('A Multiseries Horizontal Stacked Bar Chart')
plt.barh(index,series1,color='w',hatch='xx')
plt.barh(index,series2,color='w',hatch='///', left=series1)
plt.barh(index,series3,color='w',hatch='\\\\\\',left=(series1+series2))
plt.yticks(index,['Jan15','Feb15','Mar15','Apr15'])
```

This code snippet creates a multiseries horizontal stacked bar chart, using hatches to distinguish between the series instead of solid colors. Here's a detailed explanation:

*   **`plt.title('A Multiseries Horizontal Stacked Bar Chart')`**: This sets the title of the plot.
*   **`plt.barh(index,series1,color='w',hatch='xx')`**: This creates the first (bottom) layer of the horizontal stacked bars:
    *   `index`: The y-coordinates where the bars will be placed.

    *   `left=series1`: This is crucial for horizontal stacking, indicating that these bars start at the end of `series1` bars.

*   **`plt.barh(index,series3,color='w',hatch='\\\',left=(series1+series2))`**: This creates the third layer:

    *   `hatch='\\\'`: Fills these bars with a backslash pattern. Note the double backslashes are needed to escape the backslash character itself in a Python string.

#Other Bar Chart Representations

Another type of very useful representation is that of a bar chart for comparison, where two series of values sharing the same categories are compared by placing the bars in opposite directions along the y axis. In
order to do this, you have to put the y values of one of the two series in a negative form. Also in this example,
you will see the possibility of coloring the edge of the bars and their inner color in a different way. In fact, you can do this by setting the two different colors on two specific kwargs: facecolor and edgecolor.

Furthermore, in this example, you will see how to add the y value with a label at the end of each bar. This could be useful to increase the readability of the bar chart. You can do this using a for loop in which
the text() function will show the y value. You can adjust the label position with the two kwargs ha and va, which control the horizontal and vertical alignment respectively. The result will be the chart shown in the Figure

```python
import matplotlib.pyplot as plt
import numpy as np
x0 = np.arange(8)
y1 = np.array([1,3,4,6,4,3,2,1])
y2 = np.array([1,2,5,4,3,3,2,1]) # Corrected Y2 to y2
plt.ylim(-7,7)
plt.bar(x0,y1,0.9,facecolor='r',edgecolor='w')
plt.bar(x0,-y2,0.9,facecolor='b',edgecolor='w')
plt.xticks(())
plt.grid(True)
for x, y_val in zip(x0, y1): # Changed 'y' to 'y_val' to avoid conflict with the y variable from previous cells
  plt.text(x + 0.4, y_val + 0.05, '%d' % y_val, ha='center', va= 'bottom')
for x, y_val in zip(x0, y2): # Changed 'y' to 'y_val' to avoid conflict with the y variable from previous cells
  plt.text(x + 0.4, -y_val - 0.05, '%d' % y_val, ha='center', va= 'top')
```

This code for the comparative bar chart executed successfully, and a plot has been generated. Here's a detailed explanation of the code:

*   **`import matplotlib.pyplot as plt`** and **`import numpy as np`**: These lines import the necessary libraries. `matplotlib.pyplot` is used for creating plots, and `numpy` is used for numerical operations.

*   **`x0 = np.arange(8)`**: This creates a NumPy array `[0, 1, 2, 3, 4, 5, 6, 7]`. This array serves as the x-coordinates for the bars.

*   **`y1 = np.array([1,3,4,6,4,3,2,1])`** and **`y2 = np.array([1,2,5,4,3,3,2,1])`**: These lines define two NumPy arrays, `y1` and `y2`, which hold the data values for the two sets of bars being compared.

*   **`plt.ylim(-7,7)`**: This sets the y-axis limits of the plot from -7 to 7. This is important to ensure both the positive and negative bars are displayed appropriately.

*   **`plt.bar(x0,y1,0.9,facecolor='r',edgecolor='w')`**: This plots the first set of bars (`y1`):
    *   `x0`: The x-coordinates for the bars.
    *   `y1`: The heights of the bars.
    *   `0.9`: The width of each bar.
    *   `facecolor='r'`: Sets the inner color of the bars to red.
    *   `edgecolor='w'`: Sets the color of the bar edges to white.

*   **`plt.bar(x0,-y2,0.9,facecolor='b',edgecolor='w')`**: This plots the second set of bars (`y2`) for comparison:
    *   `x0`: The x-coordinates for the bars.
    *   `-y2`: The negative of the `y2` values, which makes these bars extend downwards from the x-axis.
    *   `0.9`: The width of each bar.
    *   `facecolor='b'`: Sets the inner color of these bars to blue.
    *   `edgecolor='w'`: Sets the color of their edges to white.

*   **`plt.xticks(())`**: This removes the default x-axis tick labels, as specific categories are not being labeled here.

*   **`plt.grid(True)`**: This adds a grid to the plot, making it easier to read the values.

*   **`for x, y_val in zip(x0, y1): plt.text(x + 0.4, y_val + 0.05, '%d' % y_val, ha='center', va= 'bottom')`**: This loop adds text labels (the value of `y1`) at the end of each positive bar.
    *   `x + 0.4`: Adjusts the x-position of the text to be centered on the bar (since the bar width is 0.9, 0.4 roughly corresponds to the center).
    *   `y_val + 0.05`: Places the text slightly above the bar.
    *   `'%d' % y_val`: Formats the value as an integer string.
    *   `ha='center'` and `va='bottom'`: Horizontally centers the text and vertically aligns it to the bottom, placing it just above the bar.

*   **`for x, y_val in zip(x0, y2): plt.text(x + 0.4, -y_val - 0.05, '%d' % y_val, ha='center', va= 'top')`**:
    This loop does the same for the negative bars, placing the text labels slightly below each bar.
    *   `-y_val - 0.05`: Places the text slightly below the negative bar.
    *   `ha='center'` and `va='top'`: Horizontally centers the text and vertically aligns it to the top, placing it just below the bar.

#Pie Charts

An alternative way to display data to the bar charts is the pie chart, easily obtainable using the pie() function.

Even for this type of function, you pass as the main argument a list containing the values to be displayed. I chose the percentages (their sum is 100) but actually you can use any kind of value. It will be the pie() function to inherently calculate the percentage occupied by each value.

Also with this type of representation, you need to define some key features making use of the kwargs. For example, if you want to define the sequence of the colors, which will be assigned to the sequence of input values correspondingly, you have to use the colors kwarg. Therefore you have to assign a list of strings, each containing the name of the desired color. Another important feature is to add labels to each slice of the pie. To do this, you have to use the labels kwarg to which you will assign a list of strings containing the labels to be displayed in sequence.

In addition, in order to draw the pie chart in a perfectly spherical way you have to add the axis() function at the end, specifying the string ‘equal’ as an argument. You will get a pie chart as shown in Figure

```python
import matplotlib.pyplot as plt
labels = ['Nokia','Samsung','Apple','Lumia']
values = [10,30,45,15]
colors = ['yellow','green','red','blue']
plt.pie(values,labels=labels,colors=colors)
plt.axis('equal')
```

This code snippet uses Matplotlib to create and display a pie chart. Let's break it down:

*   **`import matplotlib.pyplot as plt`**: This line imports the `pyplot` module from the `matplotlib` library, commonly aliased as `plt` for convenience. It's the standard way to bring Matplotlib's plotting functionalities into your Python script.
*   **`labels = ['Nokia','Samsung','Apple','Lumia']`**: This creates a Python list named `labels` containing strings. These strings will serve as the labels for each slice of the pie chart.
*   **`values = [10,30,45,15]`**: This creates a Python list named `values` containing numerical data. These numbers represent the relative size of each slice. Matplotlib will automatically calculate the percentage each value represents out of the total sum.
*   **`colors = ['yellow','green','red','blue']`**: This creates a Python list named `colors` containing strings, each representing a color. These colors will be applied to the corresponding pie slices in the order they appear.
*   **`plt.pie(values,labels=labels,colors=colors)`**: This is the core function call that generates the pie chart:
    *   `values`: Provides the data for the pie slices.
    *   `labels=labels`: Assigns the labels list to each slice.
    *   `colors=colors`: Assigns the colors list to each slice.
*   **`plt.axis('equal')`**: This function ensures that the pie chart is drawn as a perfect circle rather than an ellipse. Without it, Matplotlib might distort the aspect ratio based on the figure size, making the pie look squashed.

To add complexity to the pie chart, you can draw it with a slice extracted from the pie. This is usually done when you want to focus on a specific slice. In this case, for example, you would highlight the slice
referring to Nokia. In order to do this there is a special kwarg named explode. It is nothing but a sequence of float values of 0 or 1, where 1 corresponds to the fully extended slice and 0 corresponds to slices completely in the pie. All intermediate values correspond to an intermediate degree of extraction (see Figure).

You can also add a title to the pie chart with the title() function. You can also adjust the angle of rotation of the pie by adding the startangle kwarg that takes an integer value between 0 and 360, which are the degrees of rotation precisely (0 is the default value).

The modified chart should appear as in Figure

```python
import matplotlib.pyplot as plt
labels = ['Nokia','Samsung','Apple','Lumia']
values = [10,30,45,15]
colors = ['yellow','green','red','blue']
explode = [0.2,0,0,0]
plt.title('A Pie Chart')
plt.pie(values,labels=labels,colors=colors,explode=explode,startangle=180)
plt.axis('equal')
```

This code snippet uses Matplotlib to create and display a pie chart. Let's break it down:

*   **`import matplotlib.pyplot as plt`**: This line imports the `pyplot` module from the `matplotlib` library, commonly aliased as `plt` for convenience. It's the standard way to bring Matplotlib's plotting functionalities into your Python script.
*   **`labels = ['Nokia','Samsung','Apple','Lumia']`**: This creates a Python list named `labels` containing strings. These strings will serve as the labels for each slice of the pie chart.
*   **`values = [10,30,45,15]`**: This creates a Python list named `values` containing numerical data. These numbers represent the relative size of each slice. Matplotlib will automatically calculate the percentage each value represents out of the total sum.
*   **`colors = ['yellow','green','red','blue']`**: This creates a Python list named `colors` containing strings, each representing a color. These colors will be applied to the corresponding pie slices in the order they appear.
*   **`explode = [0.3,0,0,0]`**: This list controls which slices "explode" or are pulled out from the center of the pie. A value of 0.3 for the first element means the first slice (Nokia) will be pulled out by 30% of the radius, while 0 for the others means they remain attached.
*   **`plt.title('A Pie Chart')`**: This sets the title of the pie chart to "A Pie Chart".
*   **`plt.pie(values,labels=labels,colors=colors,explode=explode,startangle=180)`**: This is the core function call that generates the pie chart:
    *   `values`: Provides the data for the pie slices.
    *   `labels=labels`: Assigns the labels list to each slice.
    *   `colors=colors`: Assigns the colors list to each slice.
    *   `explode=explode`: Applies the explosion effect defined by the `explode` list.
    *   `startangle=180`: Rotates the start of the first slice by 180 degrees counter-clockwise from the x-axis. By default, the first slice starts at 0 degrees.
*   **`plt.axis('equal')`**: This function ensures that the pie chart is drawn as a perfect circle rather than an ellipse. Without it, Matplotlib might distort the aspect ratio based on the figure size, making the pie look squashed.

But the possible additions that you can insert in a pie chart do not end here. For example, a pie chart does not have axes with ticks and so it is difficult to imagine the perfect percentage represented by each slice. To overcome this, you will use the autopct kwarg that adds to the center of each slice a text label which
shows the corresponding value.

If you wish to make it an even more appealing image, you can add a shadow with the shadow kwarg setting it to True. In the end you will get a pie chart as shown in Figure

```python
import matplotlib.pyplot as plt
labels = ['Nokia','Samsung','Apple','Lumia']
values = [10,30,45,15]
colors = ['yellow','green','red','blue']
explode = [0.3,0,0,0]
plt.title('A Pie Chart')
plt.pie(values,labels=labels,colors=colors,explode=explode,
shadow=True,autopct='%1.1f%%',startangle=180)
plt.axis('equal')
```

#Contour Plot

A quite common type of chart in the scientific world is the contour plot or contour map. This visualization is in fact suitable for displaying three-dimensional surfaces through a contour map composed of curves closed showing the points on the surface that are located at the same level, or that have the same z value.

Although visually the contour plot is a very complex structure, its implementation is not so difficult, thanks to the matplotlib library. First, you need the function z = f (x, y) for generating a three-dimensional surface. Then, once you have defined a range of values x, y that will define the area of the map to be displayed, you can calculate the z values for each pair (x, y), applying the function f(x, y) just defined in order to obtain a matrix of z values. Finally, thanks to the contour() function you can generate the contour map of the surface. It is often desirable to add also a color map along with a contour map. That is, the areas delimited by the curves of level are filled by a color gradient, defined by a color map. For example, as in Figure, you may indicate negative values with increasingly dark shades of blue, and move to yellow and
then red with the increase of the positive values.

```python
import matplotlib.pyplot as plt
import numpy as np
dx = 0.01; dy = 0.01
x = np.arange(-2.0,2.0,dx)
y = np.arange(-2.0,2.0,dy)
X,Y = np.meshgrid(x,y)
def f(x,y):
  return (1 - y**5 + x**5)*np.exp(-x**2-y**2)
C = plt.contour(X,Y,f(X,Y),8,colors='black')
plt.contourf(X,Y,f(X,Y),8)
plt.clabel(C, inline=1, fontsize=10)
```

This code snippet uses Matplotlib and NumPy to create and display a contour plot, which is excellent for visualizing 3D surfaces on a 2D plane. Let's break it down:

*   **`import matplotlib.pyplot as plt`**: Imports the `pyplot` module from Matplotlib, commonly aliased as `plt`, for creating plots.
*   **`import numpy as np`**: Imports the `NumPy` library, aliased as `np`, which is essential for numerical operations, especially array manipulation and mathematical functions.
*   **`dx = 0.01; dy = 0.01`**: Defines the step size for x and y coordinates. These small values ensure a smooth contour.
*   **`x = np.arange(-2.0,2.0,dx)`**: Creates a NumPy array `x` ranging from -2.0 up to (but not including) 2.0, with increments of `dx` (0.01). This will be the x-axis range.
*   **`y = np.arange(-2.0,2.0,dy)`**: Similar to `x`, this creates a NumPy array `y` for the y-axis range.
*   **`X,Y = np.meshgrid(x,y)`**: This function creates 2D coordinate arrays `X` and `Y` from the 1D `x` and `y` arrays. `X` will have rows where each row is a copy of `x`, and `Y` will have columns where each column is a copy of `y`. This is crucial for evaluating a 2D function over a grid.
*   **`def f(x,y): return (1 - y**5 + x**5)*np.exp(-x**2-y**2)`**: Defines a Python function `f(x,y)` that calculates the z-value for given x and y coordinates. This function represents the 3D surface we want to visualize.
*   **`C = plt.contour(X,Y,f(X,Y),8,colors='black')`**: This is the first plotting command:

#Polar Chart

Another type of advanced chart that is having some success is the polar chart. This type of chart is characterized by a series of sectors which extend radially; each of these areas will occupy a certain angle. Thus you can display two different values assigning them to the magnitudes that characterize the polar chart: the extension of the radius r and the angle q occupied by the sector. These in fact are the polar coordinates (r, q), an alternative way of representing functions at the coordinate axes. From the graphical
point of view you could imagine it as a kind of chart that has characteristics both of the pie chart and of the bar chart. In fact as the pie chart, the angle of each sector gives percentage information represented by that category with respect to the total. As for the bar chart, the radial extension is the numerical
value of that category.

So far we have used the standard set of colors using single characters as the color code (e.g., ‘r’ to indicate red). In fact you can use any sequence of colors you want. You have to define a list of string values which contain RGB codes in the #rrggbb format corresponding to the colors you want.

Oddly, for getting a polar chart you have to use the bar() function in which you pass the list containing the angles q and a list of the radial extension of each sector. The result will be a polar chart as shown in
Figure

```python
import matplotlib.pyplot as plt
import numpy as np
N = 8
theta = np.arange(0.,2 * np.pi, 2 * np.pi / N)
radii = np.array([4,7,5,3,1,5,6,7])
plt.axes([0.025, 0.025, 0.95, 0.95], polar=True)
colors = np.array(['#4bb2c5', '#c5b47f', '#EAA228', '#579575', '#839557', '#958c12',
'#953579', '#4b5de4'])
bars = plt.bar(theta, radii, width=(2*np.pi/N), bottom=0.0, color=colors)
```

This code snippet uses Matplotlib and NumPy to create a polar bar chart. Let's break down each part:

*   **`import matplotlib.pyplot as plt`**: Imports the `pyplot` module from the `matplotlib` library, commonly aliased as `plt` for convenience. This is standard for Matplotlib plotting.
*   **`import numpy as np`**: Imports the `NumPy` library, aliased as `np`, which is essential for numerical operations, especially array creation and mathematical constants like `pi`.
*   **`N = 8`**: Defines an integer `N` as 8. This represents the number of bars (or sectors) in the polar chart.
*   **`theta = np.arange(0.,2 * np.pi, 2 * np.pi / N)`**: This line creates a `NumPy` array `theta`.
    *   `np.arange(start, stop, step)` generates a sequence of numbers. Here, it starts from 0., goes up to (but not including) `2 * np.pi` (a full circle in radians), with a step size determined by `(2 * np.pi / N)`. This effectively divides the full circle into `N` equal angles, which will be the positions of the bars.
*   **`radii = np.array([4,7,5,3,1,5,6,7])`**: This creates a `NumPy` array `radii` containing 8 numerical values. These values will determine the length (radius) of each bar in the polar chart.
*   **`plt.axes([0.025, 0.025, 0.95, 0.95], polar=True)`**: This creates a new set of axes for the plot.
    *   `[0.025, 0.025, 0.95, 0.95]` specifies the position and size of the axes within the figure as a fraction of the figure width and height (left, bottom, width, height).
    *   `polar=True` is the crucial argument that tells Matplotlib to create polar axes, meaning the plot will use radial and angular coordinates instead of Cartesian (x, y) coordinates.
*   **`colors = np.array(['#4bb2c5', '#c5b47f', '#EAA228', '#579575', '#839557', '#958c12', '#953579', '#4b5de4'])`**: This creates a `NumPy` array `colors` containing hexadecimal color codes. Each code corresponds to the color of one of the 8 bars.
*   **`bars = plt.bar(theta, radii, width=(2*np.pi/N), bottom=0.0, color=colors)`**: This is the main command to create the polar bar chart:
    *   `theta`: Provides the angular positions of the bars.
    *   `radii`: Provides the height (radial length) of each bar.
    *   `width=(2*np.pi/N)`: Sets the angular width of each bar, making them fit perfectly around the circle with no gaps (as it's the same as the step size for `theta`).
    *   `bottom=0.0`: Specifies the radial coordinate from which the bars extend. Here, they start from the center (radius 0).
    *   `color=colors`: Assigns the custom `colors` array to color each bar individually.

In this example you have defined the sequence of colors using the format #rrggbb, but you can specify a sequence of colors as strings with their actual name

```python
import matplotlib.pyplot as plt
import numpy as np
N = 8
theta = np.arange(0.,2 * np.pi, 2 * np.pi / N)
radii = np.array([4,7,5,3,1,5,6,7])
plt.axes([0.025, 0.025, 0.95, 0.95], polar=True)
colors = np.array(['lightgreen', 'darkred', 'navy', 'brown', 'violet', 'plum',
'yellow', 'darkgreen'])
bars = plt.bar(theta, radii, width=(2*np.pi/N), bottom=0.0, color=colors)
```

# 3D Surfaces

In the previous section you used the contour plot to represent the three-dimensional surfaces through the
level lines. Using the mplot3D package, surfaces can be drawn directly in 3D. In this example you will use again the same function z = f (x, y) you have used in the contour map.

Once you have calculated the meshgrid you can view the surface with the plot_surface() function. A three-dimensional surface of blue color will appear as in Figure

```python
from mpl_toolkits.mplot3d import Axes3D
import matplotlib.pyplot as plt
import numpy as np

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d') # Use add_subplot with projection for a clearer setup

X = np.arange(-2,2,0.1)
Y = np.arange(-2,2,0.1)
X,Y = np.meshgrid(X,Y)

def f(x,y):
  return (1 - y**5 + x**5)*np.exp(-x**2-y**2)

ax.plot_surface(X,Y,f(X,Y), rstride=1, cstride=1, cmap='viridis') # Added cmap

ax.set_title('3D Surface Plot of f(x,y)') # Added title
ax.set_xlabel('X-axis') # Added X-axis label
ax.set_ylabel('Y-axis') # Added Y-axis label
ax.set_zlabel('Z-axis') # Added Z-axis label

plt.show() # Added to display the plot
```

This code generates a 3D surface plot of the function $f(x,y) = (1 - y^5 + x^5) * e^{(-x^2 - y^2)}$.

This code snippet demonstrates how to create a complex grid of subplots using `matplotlib.gridspec.GridSpec`.

1.  **`import matplotlib.pyplot as plt`**: Imports the `pyplot` module for plotting.
2.  **`gs = plt.GridSpec(3,3)`**: Creates a `GridSpec` object that defines a 3x3 grid for placing subplots.
3.  **`fig = plt.figure(figsize=(6,6))`**: Creates a new figure with a specified size.
4.  **`fig.add_subplot(gs[1,:2])`**: Adds a subplot to the figure. `gs[1,:2]` means this subplot will span across the second row (index 1) and the first two columns (indices 0 and 1) of the 3x3 grid.
5.  **`fig.add_subplot(gs[0,:2])`**: Adds another subplot, spanning the first row (index 0) and the first two columns.
6.  **`fig.add_subplot(gs[2,0])`**: Adds a subplot to the third row (index 2) and the first column (index 0).
7.  **`fig.add_subplot(gs[:2,2])`**: Adds a subplot spanning the first two rows (indices 0 and 1) and the third column (index 2).
8.  **`fig.add_subplot(gs[2,1:])`**: Adds a final subplot spanning the third row (index 2) and the second and third columns (indices 1 and 2).

This setup allows for highly flexible arrangements of subplots with varying sizes and positions within a single figure, going beyond the simple `plt.subplot()` or `plt.subplots()` functions.
