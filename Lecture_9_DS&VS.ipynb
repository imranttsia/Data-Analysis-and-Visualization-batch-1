#Filtering Out NaN Values

There are various options to eliminate the NaN values during the data analysis. However, the elimination by hand, element by element, can be very tedious and risky, because you never get the certainty of having eliminated all the NaN values. The dropna( ) function comes to your aid.

````python
ser
````

````python
ser.dropna()
````

Another possibility is to directly perform the filtering function by placing the notnull( ) in the selection condition.

````python
ser.notnull()
````

````python
ser[ser.notnull()]
````

If you’re dealing with the DataFrame it gets a little more complex. If you use the dropna( ) function on this type of object, it is sufficient that there is only one value NaN on a column or a row to eliminate it completely.

````python
frame3 = pd.DataFrame([[6,np.nan,6],[np.nan,np.nan,np.nan],[2,np.nan,5]],
... index = ['blue','green','red'],
... columns = ['ball','mug','pen'])
frame3
````

Therefore to avoid having entire rows and columns disappear completely, you should specify the how
option, assigning a value of ‘all’ to it, in order to inform the dropna( ) function to delete only the rows or columns in which all elements are NaN.

````python
frame3.dropna()
````

````python
frame3.dropna(how='all')
````

````python
frame3.dropna(how='all',axis=1)
````

#Hierarchical Indexing and Leveling

The hierarchical indexing is a very important feature of pandas, as it allows you to have multiple levels of indexes on a single axis. Somehow it gives you a way to work with data in multiple dimensions continuing to work in a two-dimensional structure.

You can start with a simple example, creating a series containing two arrays of indexes, that is, creating a structure with two levels.

````python
mser = pd.Series(np.random.rand(8),
... index=[['white','white','white','blue','blue','red','red','red'],
... ['up','down','right','up','down','up','down','left']])
mser
````

````python
mser.index
````

Through the specification of a hierarchical indexing, the selection of subsets of values is in a certain way simplified.

In fact, you can select the values for a given value of the first index, and you do it in the classic way:

````python
mser['white']
````

````python
#or you can select values for a given value of the second index, in the following manner:
mser[:,'up']
````

````python
#Intuitively, if we want to select a specific value, you will specify both indexes.
mser['white','up']
````

The hierarchical indexing plays a critical role in reshaping the data and group-based operations such
as creating a pivot-table. For example, the data could be used just rearranged in a data frame using a special function called unstack( ). This function converts the Series with hierarchical index in a simple DataFrame where the second set of indexes is converted into a new set of columns.

````python
mser.unstack()
````

If what we want is to perform the reverse operation, which is to convert a DataFrame in a Series, you will use the stack( ) function.

#Reordering and Sorting Levels

Occasionally, you could need to rearrange the order of the levels on an axis or do a sorting for values at a specific level.

The swaplevel( ) function accepts as argument the names assigned to the two levels that you want to interchange, and returns a new object with the two levels interchanged between them, while leaving the data
unmodified.

````python
mframe.columns.names = ['objects','id']
mframe.index.names = ['colors','status']
mframe
````

````python
mframe.swaplevel('colors','status')
````

````python
mframe.sort_index(level='colors')
````

````python
mframe.sort_index(level=1)
````

### Summary:

### Pandas DataFrame Key Findings

*   **DataFrame Creation**: DataFrames can be created from various data sources.
    *   **From Dictionaries**: A dictionary where keys are column names and values are lists (or Series) of data forms a DataFrame. Pandas automatically assigns a default `RangeIndex` (0, 1, 2, ...) if no index is specified.
    *   **Specifying Columns**: When creating from a dictionary, specific columns can be selected and ordered using the `columns` parameter.
    *   **Custom Index**: A custom `index` can be provided as a list of labels during DataFrame creation, similar to Series.
    *   **From NumPy Arrays**: A DataFrame can be constructed from a NumPy array, providing `index` (for rows) and `columns` (for columns) as separate parameters. This is useful for creating DataFrames with structured numerical data.
    *   **From Nested Dictionaries**: Nested dictionaries can be directly converted into DataFrames, where the outer keys become column names and inner keys become row indices, demonstrating a powerful way to handle semi-structured data. Missing values are filled with `NaN`.

*   **Element Selection**: Accessing data within DataFrames is highly flexible.
    *   **Column Selection**: Individual columns can be selected using dictionary-like notation (`frame['column_name']`) or attribute access (`frame.column_name`), returning a `Series`.
    *   **Row Selection (Integer-Location)**: Rows can be selected by their integer position using `.iloc[]`, returning a `Series` (if a single row) or a `DataFrame` (if multiple rows/slice).
    *   **Specific Cell Selection**: A single value can be accessed by first selecting the column and then the row index (`frame['column_name'][row_index]`).

*   **Value Assignment**: DataFrames allow modification and expansion.
    *   **Index and Column Naming**: `frame.index.name` and `frame.columns.name` can be used to assign descriptive names to the row and column indices.
    *   **Adding New Columns**: New columns can be added by assigning a scalar value (which broadcasts to all rows) or a `Series` (which aligns by index) to a new column name.
    *   **Updating Existing Columns/Cells**: Existing columns can be updated by assigning a new `Series` or list. Individual cell values can be updated using combined indexing (e.g., `frame['price'][2] = 3.3`).

*   **Membership and Filtering**:
    *   **`isin()` Method**: The `isin()` method checks for the presence of values within the DataFrame, returning a boolean DataFrame. This boolean DataFrame can then be used to filter the original DataFrame, showing only matching values and `NaN` for non-matching ones.
    *   **Conditional Filtering**: Rows can be filtered based on conditions applied to column values (e.g., `frame[frame['price'] < 1]`), returning a subset of the DataFrame.

*   **Deleting Columns and Rows**:
    *   **Deleting Columns**: The `del` keyword can permanently remove a column from a DataFrame (`del frame['new']`).
    *   **Dropping Rows/Columns with `drop()`**: The `drop()` method removes rows or columns specified by their labels. For columns, `axis=1` must be specified. It returns a new DataFrame without the dropped items.

*   **Transposition**:
    *   The `.T` attribute provides a simple way to transpose a DataFrame, swapping rows and columns.

*   **Index Objects**:
    *   **Immutability**: Index objects are immutable, ensuring their integrity and safe sharing across data structures.
    *   **`is_unique` Attribute**: Checks if all labels in an index are unique.
    *   **`idxmin()` and `idxmax()`**: Return the labels of the minimum and maximum values in a Series.

*   **Reindexing**:
    *   **`reindex()` Function**: Creates a new Series or DataFrame with a new index, reordering existing data and introducing `NaN` for new labels.
    *   **Filling Missing Values during Reindexing**: The `method` parameter in `reindex()` (e.g., `'ffill'` for forward fill, `'bfill'` for backward fill) can be used to fill `NaN` values created by new indices with existing data.

*   **Arithmetic Operations and Data Alignment**:
    *   **Between Series**: When performing arithmetic operations (e.g., `s1 + s2`) between two Series, pandas automatically aligns data based on common labels. Labels present in only one Series result in `NaN` in the output.
    *   **Between DataFrames**: Similarly, operations between DataFrames (`frame1 + frame2`) align both row and column labels. Non-matching labels in either axis result in `NaN`.

*   **Flexible Arithmetic Methods**:
    *   Methods like `.add()`, `.sub()`, `.div()`, and `.mul()` offer the same functionality as their operator counterparts (e.g., `frame1.add(frame2)` is equivalent to `frame1 + frame2`).
    *   **`fill_value` Parameter**: These methods include a `fill_value` parameter to specify a value to use for `NaN` entries during alignment, preventing `NaN` propagation for non-matching labels.

*   **Operations between DataFrame and Series**:
    *   Arithmetic operations between a DataFrame and a Series (e.g., `frame - ser`) align the Series' index with the DataFrame's columns by default, applying the Series operation across each row of the DataFrame. If the Series has labels not present in the DataFrame's columns, new columns are created with `NaN` values.

*   **Function Application**:
    *   **NumPy Universal Functions (`ufuncs`)**: NumPy functions like `np.sqrt()` can be directly applied to DataFrames, performing element-wise operations.
    *   **`apply()` Method**: Custom functions (lambda or user-defined) can be applied to DataFrames.
        *   Applied by default column-wise, returning a `Series`.
        *   Can be applied row-wise by setting `axis=1`.
        *   Functions can return a `Series` to generate new DataFrames (e.g., min/max for each column).

*   **Statistical Functions**:
    *   DataFrames have built-in methods for common descriptive statistics, such as `mean()`, `sum()`, and `describe()`. By default, these operate column-wise, but can be adjusted to operate row-wise using the `axis` parameter (e.g., `frame.mean(axis=1)`).

*   **Sorting and Ranking**:
    *   **`sort_index()`**: Sorts the DataFrame or Series by its index labels. Can be applied to rows or columns (`axis=1`) and in ascending or descending order (`ascending=False`).
    *   **`sort_values()`**: Sorts a DataFrame by the values of one or more specified columns (`by` parameter). For Series, it sorts by the values directly.
    *   **`rank()`**: Assigns a rank to each element in a Series, based on its value. Can be configured with `method` (e.g., `'first'`) and `ascending` parameters.

*   **Correlation and Covariance**:
    *   **`corr()` and `cov()`**: For two Series, these functions calculate their correlation and covariance. For a DataFrame, they return a correlation or covariance matrix between all pairs of columns.
    *   **`corrwith()`**: Calculates pairwise correlations of columns or rows of a DataFrame with a Series or another DataFrame.

*   **NaN Value Handling**:
    *   **Explicit Assignment**: `np.nan` or `None` can be explicitly assigned to introduce missing values.
    *   **Identification**: `isnull()` and `notnull()` methods identify missing or non-missing values, respectively, returning boolean Series or DataFrames.
    *   **Removal (`dropna()`)**:
        *   `dropna()` removes rows/columns containing `NaN` values.
        *   The `how='all'` parameter specifies removal only if *all* elements in a row/column are `NaN`.
    *   **Filling (`fillna()`)**:
        *   `fillna()` replaces `NaN` values with a specified scalar value.
        *   Can also replace `NaN`s with different values per column by passing a dictionary mapping column names to fill values.

*   **Hierarchical Indexing and Leveling**:
    *   **MultiIndex Creation**: DataFrames and Series can have multiple levels of indices on one or both axes, creating a `MultiIndex`.
    *   **Selection with MultiIndex**: Data can be selected at different levels of a MultiIndex (e.g., `mser['white']` or `mser[:,'up']`).
    *   **`unstack()` and `stack()`**:
        *   `unstack()` pivots the innermost (or specified) index level to become a new column level, converting a Series with a MultiIndex into a DataFrame.
        *   `stack()` performs the reverse operation, converting columns into an inner index level, turning a DataFrame into a Series with a MultiIndex.
    *   **Reordering Levels (`swaplevel()`)**: The order of index levels can be interchanged without altering the data.
    *   **Sorting Levels (`sort_index(level=...)`)**: Data can be sorted based on a specific level of a MultiIndex.
    *   **Summary Statistics by Level (`groupby(level=...).sum()`)**: Statistical operations can be applied to specific levels of a MultiIndex, aggregating data across those levels.
