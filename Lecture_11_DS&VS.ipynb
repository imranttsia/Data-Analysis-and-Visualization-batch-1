#PANDAS in Depth: Data Manipulation

Once you have the data in DataFrame format, they are ready to be manipulated. The manipulation of the data has the purpose of preparing the data so that they can be more easily subjected to analysis. In fact, their
manipulation will depend a lot on purposes of those who must carry out the analysis, and it will be performed for making more explicit the information you are looking for. Especially in preparation for the next phase, the
data must be ready to the data visualization.

The three phases of data manipulation are

• Data preparation

• Data transformation

• Data aggregation

#Data Preparation

Before you start manipulating data itself, it is necessary to prepare the data and assemble them in the form of data structures such that they can be manipulated later with the tools made available by the pandas library.

The different procedures for data preparation are listed below.

• loading

• assembling

              • merging

              • concatenating

              • combining

• reshaping (pivoting)

• removing

In the loading phase, there is also that part of the preparation which concerns the conversion from many different formats into a data structure such as DataFrame. But even after you have gotten the data, probably from different sources and formats, and unified it into a DataFrame, you will need to perform further operations of preparation.

The data contained in the pandas objects can be assembled together in different ways:

• Merging—the pandas.merge( ) function connects the rows in a DataFrame based on one or more keys.

• Concatenating—the pandas.concat( ) function concatenates the objects along an axis.

• Combining—the pandas.DataFrame.combine_first( ) function is a method that
allows you to connect overlapped data in order to fill in missing values in a data structure by taking data from another structure.

Furthermore, part of the preparation process is also pivoting, which consists of the exchange between rows and columns.

#Merging

The merging operation, which corresponds to the JOIN operation for those who are familiar with SQL, consists of a combination of data through the connection of rows using one or more keys.

In fact, anyone working with relational databases usually makes use of the JOIN query with SQL to get data from different tables using some reference values (keys) shared between them. On the basis of these keys it is possible to obtain new data in a tabular form as the result of the combination of other tables. This operation with the library pandas is called merging, and merge() is the function to perform this kind of
operation.

First, you have to import the pandas library and define two DataFrame that will serve you as examples
for this section.

```python
import numpy as np
import pandas as pd
frame1 = pd.DataFrame( {'id':['ball','pencil','pen','mug','ashtray'], 'price': [12.33,11.44,33.21,13.23,33.62]})
frame1
```

```python
frame2 = pd.DataFrame( {'id':['pencil','pencil','ball','pen'],'color': ['white','red','red','black']})
frame2
```

```python
pd.merge(frame1,frame2)
```

As you can see from the result, the returned DataFrame consists of all rows that have an ID in common between the two DataFeame. In addition to the common column, the columns from both the first and the second DataFrame are added.

In this case you used the merge( ) function without specifying any column explicitly. In fact, in most cases you need to decide which is the column on which to base the merging.

To do this, add the on option with the column name as the key for the merging.

```python
frame1 = pd.DataFrame( {'id':['ball','pencil','pen','mug','ashtray'],'color': ['white','red','red','black','green'],'brand': ['OMG','ABC','ABC','POD','POD']})
frame1
```

```python
frame2 = pd.DataFrame( {'id':['pencil','pencil','ball','pen'],'brand': ['OMG','POD','ABC','POD']})
frame2
```

Now in this case you have two DataFrame having columns with the same name. So if you launch a merging you do not get any results.

```python
pd.merge(frame1,frame2)
```

So it is necessary to explicitly define the criterion of merging that pandas must follow, specifying the name of the key column in the on option.

```python
pd.merge(frame1,frame2,on='id')
```

```python
pd.merge(frame1,frame2,on='brand')
```

As expected, the results vary considerably depending on the criteria of merging.

Often, however, the opposite problem arises, that is, to have two DataFrames in which the key columns do not have the same name. To remedy this situation, you have to use the left_on and right_on options that
specify the key column for the first and for the second DataFrame. Now you can see an example.

```python
frame2.columns = ['brand','sid']
frame2
```

```python
pd.merge(frame1, frame2, left_on='id', right_on='sid')
```

By default, the merge( ) function performs an inner join; the keys in the result are the result of an intersection.

Other possible options are the left join, the right join, and the outer join. The outer join produces the union of all keys, combining the effect of a left join with a right join. To select the type of join you have to use
the how option.

```python
frame2.columns = ['id','brand']
pd.merge(frame1,frame2,on='id')
```

```python
frame1
```

```python
frame2
```

```python
pd.merge(frame1,frame2,on='id',how='outer')
```

```python
pd.merge(frame1,frame2,on='id',how='left')
```

```python
pd.merge(frame1,frame2,on=['id','brand'],how='outer')
```

## Explanation of `pd.merge` Operations

1.  **`pd.merge(frame1, frame2, on='id', how='outer')`**:
    *   This performs an **outer join** between `frame1` and `frame2` using the 'id' column as the key.
    *   An outer join includes all rows from both DataFrames. If an 'id' is present in one DataFrame but not the other, the corresponding columns from the missing DataFrame will be filled with `NaN` (Not a Number).
    *   For example, 'mug' and 'ashtray' from `frame1` have no matching 'id' in `frame2`, so their `brand_y` column is `NaN`. Conversely, if there were 'id's in `frame2` not in `frame1`, their `color` and `brand_x` would be `NaN`.

2.  **`pd.merge(frame1, frame2, on='id', how='left')`**:
    *   This performs a **left join** between `frame1` and `frame2` using the 'id' column.
    *   A left join includes all rows from the *left* DataFrame (`frame1` in this case) and only the matching rows from the *right* DataFrame (`frame2`).
    *   If an 'id' from `frame1` does not have a match in `frame2` (like 'mug' and 'ashtray'), the columns from `frame2` (`brand_y` here) will show `NaN` for those rows. All rows from `frame1` are preserved.

3.  **`pd.merge(frame1, frame2, on=['id', 'brand'], how='outer')`**:
    *   This is again an **outer join**, but this time the merge is performed on *two* keys: 'id' and 'brand'.
    *   For rows to be considered a match, both the 'id' *and* the 'brand' values must be identical in both DataFrames.
    *   If a combination of 'id' and 'brand' exists in `frame1` but not `frame2`, or vice-versa, that row is included with `NaN`s for the columns from the non-matching DataFrame. This creates a more granular merge criterion, often resulting in more `NaN`s if the two DataFrames don't have many exact matches across both specified columns.

## Common Use Cases for `pd.merge()`

`pd.merge()` is an incredibly powerful function for combining data from different DataFrames. Here are a few common scenarios where it proves invaluable:

1.  **Combining Customer Information with Order Details**: Enriching order records with customer demographics.
2.  **Joining Product Listings with Category Information**: Adding descriptive category names to product data.
3.  **Merging Sales Data with Regional Information**: Analyzing sales performance across different geographical regions.

### Use Case 1: Combining Customer Information with Order Details

Imagine you have one DataFrame with customer details (like `customer_id`, `name`, `city`) and another DataFrame with order details (`order_id`, `customer_id`, `product`, `amount`). You want to see which customer placed which order.

```python
import pandas as pd

# Create a DataFrame for customer information
customers_df = pd.DataFrame({
    'customer_id': [1, 2, 3, 4],
    'name': ['Alice', 'Bob', 'Charlie', 'David'],
    'city': ['New York', 'Los Angeles', 'Chicago', 'Houston']
})

# Create a DataFrame for order details
orders_df = pd.DataFrame({
    'order_id': [101, 102, 103, 104, 105],
    'customer_id': [1, 2, 1, 3, 5], # Customer ID 5 doesn't exist in customers_df
    'product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Webcam'],
    'amount': [1200, 25, 75, 300, 50]
})

print("Customers DataFrame:")
display(customers_df)

print("\nOrders DataFrame:")
display(orders_df)

# Merge customer and order data using a left join
# We use a left join to keep all customer information and add matching order details.
# Orders without a matching customer will result in NaN customer info (if an outer join was used).
# In this case, we want to see all customer data, and their corresponding orders.
merged_customer_orders = pd.merge(customers_df, orders_df, on='customer_id', how='left')

print("\nMerged Customer and Order Data (Left Join):")
display(merged_customer_orders)

# If we wanted to see only orders for existing customers, we would use an inner join
merged_customer_orders_inner = pd.merge(customers_df, orders_df, on='customer_id', how='inner')
print("\nMerged Customer and Order Data (Inner Join):")
display(merged_customer_orders_inner)
```

### Use Case 2: Joining Product Listings with Category Information

Suppose you have a DataFrame of products (`product_id`, `product_name`, `category_id`) and another DataFrame with category details (`category_id`, `category_name`, `description`). You want to add the `category_name` to your product listings.

```python
import pandas as pd

# Create a DataFrame for product listings
products_df = pd.DataFrame({
    'product_id': ['P001', 'P002', 'P003', 'P004'],
    'product_name': ['Smartphone', 'Tablet', 'Smartwatch', 'Headphones'],
    'category_id': [10, 20, 10, 30] # Category ID 30 doesn't exist in categories_df
})

# Create a DataFrame for category details
categories_df = pd.DataFrame({
    'category_id': [10, 20, 40],
    'category_name': ['Electronics', 'Computing', 'Accessories'],
    'description': ['Devices and gadgets', 'Computers and peripherals', 'Various accessories']
})

print("Products DataFrame:")
display(products_df)

print("\nCategories DataFrame:")
display(categories_df)

# Merge product and category data using a left join
# We use a left join to keep all products and add their category names.
merged_products_categories = pd.merge(products_df, categories_df, on='category_id', how='left')

print("\nMerged Products with Category Information (Left Join):")
display(merged_products_categories)

merged_products_categories2 = pd.merge(products_df, categories_df, on='category_id', how='inner')
print("\nMerged Products with Category Information (Inner Join):")
display(merged_products_categories2)

merged_products_categories3 = pd.merge(products_df, categories_df, on='category_id', how='outer')
print("\nMerged Products with Category Information (Outer Join):")
display(merged_products_categories3)
```

### Use Case 3: Merging Sales Data with Regional Information

Consider having sales records (`sale_id`, `store_id`, `revenue`) and a separate DataFrame mapping `store_id` to `region_name`. You'd want to combine these to analyze sales by region.

```python
import pandas as pd

# Create a DataFrame for sales data
sales_df = pd.DataFrame({
    'sale_id': [501, 502, 503, 504, 505],
    'store_id': ['S1', 'S2', 'S1', 'S3', 'S4'], # Store ID S4 doesn't exist in regions_df
    'revenue': [1500, 2300, 1800, 900, 700]
})

# Create a DataFrame for store regions
regions_df = pd.DataFrame({
    'store_id': ['S1', 'S2', 'S3', 'S5'], # Store ID S5 doesn't have sales data
    'region_name': ['North', 'South', 'West', 'East']
})

print("Sales DataFrame:")
display(sales_df)

print("\nRegions DataFrame:")
display(regions_df)

# Merge sales data with regional information using an inner join
# An inner join is appropriate here if we only want to analyze sales from stores that have region information.
merged_sales_regions = pd.merge(sales_df, regions_df, on='store_id', how='inner')

print("\nMerged Sales Data with Regional Information (Inner Join):")
display(merged_sales_regions)

# If we wanted to keep all sales records and see which ones don't have region info, we'd use a left join
merged_sales_regions_left = pd.merge(sales_df, regions_df, on='store_id', how='left')
print("\nMerged Sales Data with Regional Information (Left Join):")
display(merged_sales_regions_left)
```

#Concatenating

Another type of data combination is referred to as concatenation. NumPy provides a concatenate()function to do this kind of operation with arrays.

```python
array1=np.arange(9).reshape((3,3))
array1
```

```python
array2 = np.arange(9).reshape((3,3))+6
array2
```

```python
np.concatenate([array1,array2])
```

```python
np.concatenate([array1,array2],axis=1)
```

```python
np.concatenate([array1,array2],axis=0)
```

As regards the pandas library and its data structures like Series and DataFrame, the fact of having labeled axes allows you to further generalize the concatenation of arrays. The concat() function is provided
by pandas for this kind of operation.

```python
ser1 = pd.Series(np.random.rand(4), index=[1,2,3,4])
ser1
```

```python
ser2 = pd.Series(np.random.rand(4), index=[5,6,7,8])
ser2
```

```python
pd.concat([ser1,ser2])
```

By default, the concat() function works on axis = 0, having as returned object a Series. If you set the axis = 1, then the result will be a DataFrame.

```python
pd.concat([ser1,ser2],axis=1)
```

From the result you can see that there is no overlap of data, therefore what you have just done is an outer join. This can be changed by setting the join option to ‘inner’.

```python
pd.concat([ser1,ser2],axis=0,join='inner')
```

A problem in this kind of operation is that the concatenated parts are not identifiable in the result. For example, you want to create a hierarchical index on the axis of concatenation. To do this you have to use the keys option.

```python
pd.concat([ser1,ser2], keys=[1,2])
```

In the case of combinations between Series along the axis = 1 the keys become the column headers of the DataFrame.

```python
pd.concat([ser1,ser2], axis=1, keys=[1,2])
```

So far you have seen the concatenation applied to the Series, but the same logic can be applied to the DataFrame.

```python
frame1 = pd.DataFrame(np.random.rand(9).reshape(3,3), index=[1,2,3],columns=['A','B','C'])
frame1
```

```python
frame2 = pd.DataFrame(np.random.rand(9).reshape(3,3), index=[4,5,6], columns=['A','B','C'])
frame2
```

```python
pd.concat([frame1, frame2])
```

```python
pd.concat([frame1, frame2], axis=1)
```

#Combining

There is another situation in which there is combination of data that cannot be obtained either with merging or with concatenation. Take the case in which you want the two datasets to have indexes that overlap in their entirety or at least partially.

One applicable function to Series is combine_first(), which performs this kind of operation along with an data alignment.

```python
ser1 = pd.Series(np.random.rand(5),index=[1,2,3,4,5])
ser1
```

```python
ser2 = pd.Series(np.random.rand(4),index=[2,4,5,6])
ser2
```

```python
ser1.combine_first(ser2)
```

```python
ser2.combine_first(ser1)
```

Instead, if you want a partial overlap, you can specify only the portion of the Series you want to overlap.

```python
ser1[:3].combine_first(ser2[:3])
# This operation combines the first three elements of `ser1` (by label) with the first three elements of `ser2` (by position).
# `combine_first` prioritizes non-NaN values from the first Series (`ser1[:3]`).
# For common indices (like '2'), `ser1`'s value is kept. For indices present only in `ser1[:3]` (like '1', '3'), its values are kept.
# For indices present only in `ser2[:3]` (like '4', '5'), their values are taken from `ser2`.
# Original ser1: 1    0.682345, 2    0.860347, 3    0.536425, 4    0.474636, 5    0.809059
# Original ser2: 2    0.252664, 4    0.756949, 5    0.862642, 6    0.388343
# ser1[:3] has indices 1, 2, 3.
# ser2[:3] has indices 2, 4, 5 (from original ser2).
# Output will have: Index 1 from ser1; Index 2 from ser1 (as ser1's value is not NaN); Index 3 from ser1; Index 4 from ser2; Index 5 from ser2.
```

#Pivoting

In addition to assembling the data in order to unify the values collected from different sources, another fairly common operation is pivoting. In fact, arrangement of the values by row or by column is not always
suited to your goals. Sometimes you would like to rearrange the data carrying column values on rows or vice versa.

#Pivoting with Hierarchical Indexing

You have already seen that DataFrame can support hierarchical indexing. This feature can be exploited to rearrange the data in a DataFrame. In the context of pivoting you have two basic operations:

• stacking: rotates or pivots the data structure converting columns to rows

• unstacking: converts rows into columns

```python
frame1 = pd.DataFrame(np.arange(9).reshape(3,3), index=['white','black','red'], columns=['ball','pen','pencil'])
frame1
```

```python
#Using the stack() function on the DataFrame, you will get the pivoting of the columns in rows, thus producing a Series:
ser5=frame1.stack()
ser5
```

```python
#From this hierarchically indexed series, you can reassemble the DataFrame into a pivoted table by use of the unstack() function
ser5.unstack()
```

```python
#You can also do the unstack on a different level, specifying the number of levels or its name as the argument of the function.
ser5.unstack(0)
```

```python
ser5.unstack(1)
```
