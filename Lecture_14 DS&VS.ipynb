#Data Visualization

Data visualization is the process of translating data into a visual context, such as a map, graph, or chart. It helps people understand the significance of data by placing it in a visual context. This makes complex data more accessible, understandable, and usable. Visualizations reveal trends, patterns, and outliers that might otherwise be missed in text-based data. It's a crucial step in data analysis, allowing for quicker insights and better decision-making across various fields.

#Matplotlib

Matplotlib is a comprehensive library for creating static, animated, and interactive visualizations in Python. It is a powerful and popular tool for data visualization, offering a wide array of plotting functions to generate various types of plots like line plots, scatter plots, bar charts, histograms, and more. It is highly customizable, allowing users fine-grained control over every aspect of a plot.

### Use Cases of Data Visualization and Matplotlib

Data visualization and Matplotlib are widely used across various domains due to their ability to transform complex data into understandable visual insights. Here are some key use cases:

1.  **Business Intelligence and Analytics:**
    *   **Sales Performance:** Tracking sales trends over time, comparing performance across different regions or products, and identifying top-performing areas or periods.
    *   **Customer Behavior:** Visualizing customer demographics, purchasing patterns, and engagement metrics to inform marketing strategies.
    *   **Financial Reporting:** Creating dashboards with charts (e.g., line charts for stock prices, bar charts for revenue vs. expenses) to provide clear financial overviews.

2.  **Scientific Research and Engineering:**
    *   **Experimental Data Analysis:** Plotting experimental results to identify relationships, trends, and anomalies. For example, plotting sensor data over time or visualizing the distribution of measurements.
    *   **Simulation Results:** Visualizing output from simulations (e.g., fluid dynamics, structural stress) to understand complex physical phenomena.
    *   **Medical Imaging:** Representing medical scan data (though specialized tools are often used, Matplotlib can be a foundational step for custom visualizations).

3.  **Data Science and Machine Learning:**
    *   **Exploratory Data Analysis (EDA):** Understanding data distributions (histograms), relationships between variables (scatter plots), and identifying outliers before model building.
    *   **Model Evaluation:** Visualizing model performance metrics (e.g., ROC curves, precision-recall curves), error distributions, and prediction accuracy.
    *   **Feature Engineering:** Understanding the impact of different features on the target variable.

4.  **Healthcare and Medicine:**
    *   **Patient Data Analysis:** Tracking patient vital signs, disease progression, and treatment effectiveness over time.
    *   **Public Health:** Visualizing the spread of diseases, vaccine efficacy, and health outcomes across populations.

5.  **Education and Journalism:**
    *   **Explaining Concepts:** Creating clear and intuitive graphs to illustrate complex theories or data in textbooks, presentations, or news articles.
    *   **Data-Driven Storytelling:** Journalists use visualizations to present data-backed stories, making information more engaging and accessible to the public.

6.  **Environmental Science:**
    *   **Climate Change Analysis:** Plotting temperature anomalies, sea-level rise, and carbon emissions over decades.
    *   **Pollution Monitoring:** Visualizing pollution levels in air or water across different locations or times.

These examples highlight how Matplotlib's flexibility allows users to create a wide array of static, animated, and interactive visualizations tailored to specific needs, making it an indispensable tool for data analysis and communication.

#A Simple Interactive Chart

To get familiar with the matplotlib library and in a particular way with Pyplot, you will start creating a simple interactive chart. Using matplotlib this operation is very simple; in fact, you can achieve it using only three lines of code.

But first you need to import the pyplot package and rename it as plt.

In [1]: import matplotlib.pyplot as plt

In Python programming language the constructors generally are not necessary; everything is already implicitly defined. In fact when you import the package, the plt object with all its graphics capabilities have already been instantiated and ready to use. In fact, you simply use the plot() function to pass the values to
be plotted.

Thus, you can simply pass the values that you want to represent as a sequence of integers.

In [2]: plt.plot([1,2,3,4])

Out[2]: [<matplotlib.lines.Line2D at 0xa3eb438>]

As you can see a Line2D object has been generated. The object is a line that represents the linear trend of the points included in the chart.

Now it is all set. You just have to give the command to show the plot using the show() function.

In [3]: plt.show()

The result will be the one shown in Figure 7-5. It looks just a window, the plotting window, with a
toolbar and the plot represented within it, just as with Matlab.

```python
import matplotlib.pyplot as plt
```

```python
plt.plot([1,2,3,4])
```

If you pass only a list of numbers or an array to the plt.plot() function, matplotlib assumes it is the sequence of y values of the chart, and it associates them to the natural sequence of values x: 0,1,2,3, ... .

Generally a plot represents value pairs (x, y), so if you want to define a chart correctly, you must define two arrays, the first containing the values on the x axis and the second containing the values on the y axis.

Moreover, the plot() function can accept a third argument which describes the specifics of how you want the point to be represented on the chart.

#Set the Properties of the Plot

As you can see in above Figure, the points were represented by a blue line. In fact, if you do not specify otherwise, the plot is represented taking into account a default configuration of the plt.plot() function:

• the size of the axes matches perfectly with the range of the input data

• there is neither title nor axis labels

• there is no legend

• A blue line connecting the points is drawn

Therefore you need to change this representation to have a real plot in which each pair of values (x, y) is represented by a red dot.

If you’re working on IPython, close the window to get back again to the active prompt for entering new commands. Then you have to call back the show() function to observe the changes made to the plot.

```python
plt.plot([1,2,3,4],[1,4,9,16],'ro')
```

In Matplotlib's plt.plot() function, 'ro' is a format string. The 'r' specifies the color red, and the 'o' specifies that the data points should be marked with circles. So, 'ro' means 'red circles'. If you look at the previous output, you'll see the plot is now represented with red dots instead of a blue line.

You can define the range both on the x axis and on the y axis by defining the details of a list [xmin, xmax, ymin, ymax] and then passing it as an argument to the axis() function.

```python
plt.axis([0,5,0,20])
plt.title('My first plot')
plt.plot([1,2,3,4],[1,4,9,16],'ro')
```

you can see how the new settings made the plot more readable. In fact, the end points
of the data set are now represented within the plot rather than at the edges. Also the title of the plot is now visible at the top.

#matplotlib and NumPy

Even the Matplot library, despite being a fully graphical library, has its foundation the NumPy library. In fact, you have seen so far how to pass lists as arguments, both to represent the data and to set the extremes of the axes. Actually, these lists have been converted internally in NumPy arrays.

Therefore, you can directly enter NumPy arrays as input data. This array of data, which have been processed by pandas, can be directly used with matplotlib without further processing.

As an example now you see how it is possible to plot three different trends in the same plot. You can choose for this example the sin() function belonging to the math module. So you will need to import it. To generate points following a sinusoidal trend you will use the library NumPy.

Generate a series of points on the x axis using the arange() function, while for the values on the y axis you will use the map() function to apply the sin() function on all the items of the array (without using a for loop).

```python
import math
import numpy as np
t = np.arange(0,2.5,0.1)
y1 = list(map(math.sin,math.pi*t))
y2 = list(map(math.sin,math.pi*t+math.pi/2))
y3 = list(map(math.sin,math.pi*t-math.pi/2))
plt.plot(t,y1,'b*',t,y2,'g^',t,y3,'ys')
```

Code Explanation:

1. import math and import numpy as np: These lines import the math module for mathematical functions (like sin and pi) and the numpy library, aliased as np, which is essential for numerical operations, especially with arrays.

2. t = np.arange(0, 2.5, 0.1): This creates a NumPy array named t. np.arange() generates a sequence of numbers starting from 0, up to (but not including) 2.5, with a step of 0.1. This array will serve as the x-axis values for our plots.

3. y1 = list(map(math.sin, math.pi * t)):

math.pi * t: This performs element-wise multiplication of pi with each value in the t array.

map(math.sin, ...): The map() function applies the math.sin function to each element of the resulting array. This calculates the sine of each value.

list(...): This converts the map object (which is an iterator) into a list. This step was crucial to fix the previous error, as Matplotlib expects list-like or array-like inputs.

y1 now contains a list of y-values representing a sine wave.

4. y2 = list(map(math.sin, math.pi * t + math.pi / 2)): This is similar to y1, but it adds math.pi / 2 to the argument of the sine function. This shifts the sine wave by 90 degrees (a quarter period) to the left.

5. y3 = list(map(math.sin, math.pi * t - math.pi / 2)): Similar to y1 and y2, but it subtracts math.pi / 2 from the argument, shifting the sine wave by 90 degrees (a quarter period) to the right.

6. plt.plot(t, y1, 'b*', t, y2, 'g^', t, y3, 'ys'): This is the core plotting command. plt.plot() is called with multiple sets of (x, y, format) arguments to plot three different lines on the same graph:

(t, y1, 'b*'): Plots t against y1 using blue asterisks (b*).
(t, y2, 'g^'): Plots t against y2 using green upward-pointing triangles (g^).
(t, y3, 'ys'): Plots t against y3 using yellow squares (ys).

As you can see in the plot represents the three different temporal trends with three different colors and markers. In these cases, when the trend of a function is so obvious, the plot is perhaps not the most appropriate representation, but it is better to use the lines. To differentiate the three trends as well as using different colors, you can use the pattern composed of different combinations of dots
and dashes ( - and . ).

```python
plt.plot(t,y1,'b--',t,y2,'g',t,y3,'r-.')
```

#Using the kwargs

The objects that make up a chart have many attributes that characterize them. These attributes are all default values, but can be set through the use of keyword args, often referred as kwargs.

These keywords are passed as arguments to functions called. In reference documentation of the various functions of the matplotlib library, you will always find them referred to as kwargs in the last position.

For example the plot() function that you are using in our examples is referred to in the following way.

matplotlib.pyplot.plot(*args, **kwargs)

For a practical example, the thickness of a line can be changed then if you set the linewidth keyword

```python
import matplotlib.pyplot as plt

plt.plot([1,2,4,2,1,0,1,2,1,4],linewidth=2.0)
```

#Working with Multiple Figures and Axes

So far you have seen how all pyplot commands are routed to the display of a single figure. Actually, matplotlib allows management of multiple figures simultaneously, and within each figure, it offers the ability to view different plots defined as subplots.

So when you are working with pyplot, you must always keep in mind the concept of current Figure and current Axes (that is, the plot shown within the figure).

Now you will see an example where two subplots are represented in a single figure. The subplot() function, in addition to subdividing the figure in different drawing areas, is used at the same time in order to focus the commands on a specific subplot.

The argument passed to the subplot() function sets the mode of subdivision and which is the current subplot. The current subplot will be the only figure which will be affected by the commands. The argument
of the subplot() function is composed of three integers. The first number defines how many parts the figure is split into vertically. The second number defines how many parts the figure is divided into horizontally.

The third issue selects which is the current subplot on which you can direct commands.

Now you will display two sinusoidal trends (sine and cosine) and the best way to do that is to divide the canvas vertically in two horizontal subplots (as shown in Figure 7-12). So the numbers to pass as an
argument are ‘211’ and ‘212’ .

```python
import numpy as np
t = np.arange(0,5,0.1)
y1 = np.sin(2*np.pi*t)
y2 = np.sin(2*np.pi*t)
```

```python
plt.subplot(211)
plt.plot(t,y1,'b-.')
plt.subplot(212)
plt.plot(t,y2,'r--')
```

Now you do the same thing by dividing the figure in two vertical subplots. The numbers to be passed as arguments to the subplot() function are ‘121’ and ‘122’

```python
t = np.arange(0.,1.,0.05)
y1 = np.sin(2*np.pi*t)
y2 = np.cos(2*np.pi*t)
```

```python
plt.subplot(121)
plt.plot(t,y1,'b-.')
plt.subplot(122)
plt.plot(t,y2,'r--')
```
