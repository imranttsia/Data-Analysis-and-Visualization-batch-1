```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set up the matplotlib figure
plt.figure(figsize=(12, 10))

# List of numerical columns to plot
numerical_cols = ['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']

# Loop through each numerical column and create a histogram
for i, col in enumerate(numerical_cols):
    plt.subplot(2, 2, i + 1)  # 2 rows, 2 columns, current subplot
    sns.histplot(data[col], kde=True) # kde=True adds a Kernel Density Estimate plot
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')

plt.tight_layout() # Adjust layout to prevent overlapping titles/labels
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

print('Visualizing relationships between numerical features, colored by Species:')
sns.pairplot(data.drop('Id', axis=1), hue='Species', markers=['o', 's', 'D'])
plt.suptitle('Pair Plot of Iris Features by Species', y=1.02) # Add a title above the plots
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(16, 12))

# List of numerical columns
numerical_cols = ['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']

# Loop through each numerical column and create a box plot grouped by Species
for i, col in enumerate(numerical_cols):
    plt.subplot(2, 2, i + 1) # 2 rows, 2 columns, current subplot
    sns.boxplot(x='Species', y=col, data=data, palette='viridis')
    plt.title(f'{col} by Species')
    plt.xlabel('Species')
    plt.ylabel(col)
    plt.xticks(rotation=45, ha='right') # Rotate x-axis labels for better readability

plt.tight_layout()
plt.show()
```
