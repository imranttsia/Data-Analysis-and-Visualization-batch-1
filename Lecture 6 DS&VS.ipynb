#Introduction to pandas Data Structures

The heart of pandas is just the two primary data structures on which all transactions, which are generally made during the analysis of data, are centralized:

• Series

• DataFrame

The Series, as you will see, constitutes the data structure designed to accommodate a sequence of
one-dimensional data, while the DataFrame, a more complex data structure, is designed to contain cases with several dimensions.

Although these data structures are not the universal solution to all the problems, they do provide a valid and robust tool for most  applications. In fact, in their simplicity, they remain very simple to understand and use. In addition, many cases of more complex data structures can still be traced to these simple two cases.


#The Series

The Series is the object of the pandas library designed to represent one-dimensional data structures, similarly to an array but with some additional features. Its internal structure is simple and is composed of two arrays associated with each other. The main array has the purpose to hold the data (data of any NumPy type) to which each element is associated with a label, contained within the other array, called the Index.

```python
import numpy as np
a=np.array([1,2,3,4])
a
```

```python
print(type(a))
```

```python
b=[1,2,3,4]
b
```

```python
import pandas as pd
# Declaring a Series
#To create the Series as specified in Figure 4-1, simply call the Series( ) constructor passing as an argument an array containing the values to be included in it.
s = pd.Series([12,-4,7,9])
s
```

```python
print(type(s))
```

As you can see from the output of the Series, on the left there are the values in the Index, which is a series of labels, and on the right the corresponding values.

If you do not specify any index during the definition of the Series, by default, pandas will assign numerical values increasing from 0 as labels. In this case the labels correspond to the indexes (position in the array) of the elements within the Series object.

Often, however, it is preferable to create a Series using meaningful labels in order to distinguish and identify each item regardless of the order in which they were inserted into the Series.

```python
# So in this case it will be necessary, during the constructor call, to include the index option assigning an array of strings containing the labels.
s = pd.Series([12,-4,7,9], index=['a','b','c','d'])
s
```

If you want to individually see the two arrays that make up this data structure you can call the two attributes of the Series as follows: index and values.

```python
import pandas as pd

# Create a simple Series from a list
data = [10, 20, 30, 40, 50]
series_example = pd.Series(data)

print("Series created from a list:")
display(series_example)

# Create a Series with a custom index
data_with_index = [100, 200, 300]
index_labels = ['a', 'b', 'c']
series_with_custom_index = pd.Series(data_with_index, index=index_labels)

print("\nSeries with a custom index:")
display(series_with_custom_index)

# Create a Series from a dictionary
dictionary_data = {'apple': 10, 'banana': 20, 'cherry': 30}
series_from_dict = pd.Series(dictionary_data)

print("\nSeries created from a dictionary:")
display(series_from_dict)
```

#NaN Values

As you can see in the previous case we tried to run the logarithm of a negative number and received NaN as a result. This specific value NaN (Not a Number) is used within pandas data structures to indicate the presence of an empty field or not definable numerically.

Generally, these NaN values are a problem and must be managed in some way, especially during data
analysis. These data are generated especially when extracting data from some source gave some trouble,
or even when the source is a missing data.

Furthermore, as you have just seen, the NaN values can also be generated in special cases, such as calculations of logarithms of negative values, or exceptions during execution of some calculation or function. In later chapters we will see how to apply different strategies to address the problem of NaN values.
Despite their problematic nature, however, pandas allows to explicitly define and add this value in a
data structure, such as Series. Within the array containing the values you enter np.NaN wherever we want to
define a missing value.

```python
s2 = pd.Series([5,-3,np.nan,14])
s2
```

```python
#The isnull( ) and notnull( ) functions are very useful to identify the indexes without a value.
s2.isnull( )
```

```python
s2.notnull( )
```

In fact, these functions return two Series with Boolean values that contains the ‘True’ and ‘False’ values depending on whether the item is a NaN value or less. The isnull( ) function returns ‘True’ at NaN values in the Series; inversely, the notnull( ) function returns ‘True’ if they are not NaN. These functions are useful to be placed inside a filtering to make a condition.

```python
s2[s2.notnull( )]
```

```python
s2[s2.isnull( )]
```

#Series as Dictionaries

An alternative way to see a Series is to think of them as an object dict (dictionary). This similarity is also exploited during the definition of an object Series. In fact, you can create a series from a dict previously defined.

```python
mydict = {'red': 2000, 'blue': 1000, 'yellow': 500, 'orange': 1000}
myseries = pd.Series(mydict)
myseries
```

As you can see from this example, the array of index is filled with the values of the keys while the data with the corresponding values. You can also define the array indexes separately. In this case, control of correspondence between the keys of the dict and labels array of indexes will run. In case of mismatch, pandas will add value NaN.

```python
colors = ['red','yellow','orange','blue','green']
myseries = pd.Series(mydict, index=colors)
myseries
```

## Summary:

### Data Analysis Key Findings

*   **Series Declaration Flexibility**: Pandas Series can be effectively declared from various data sources, including lists of floats (e.g., `[1.1, 2.2, 3.3, 4.4]` creating a `float64` Series), lists of booleans (`[True, False, True, True]` creating a `bool` Series), scalar values (e.g., `5` replicated across a custom index), and NumPy arrays with custom indices.
*   **Comprehensive Element Selection and Assignment**:
    *   Elements can be precisely selected using label-based indexing (`.loc[]` for single or multiple labels) and integer-location based indexing (`.iloc[]` for single positions or slices).
    *   Boolean indexing allows for advanced filtering, such as selecting values greater than 5.
    *   Value assignment is versatile, supporting direct label assignment, slicing, boolean condition-based assignment, and targeted updates using `.loc[]` and `.iloc[]`.
*   **Series Creation and Data Alignment**:
    *   A Series can be created directly from a NumPy array with a custom index, providing clear label association.
    *   Creating a Series from an existing one with a redefined index demonstrates pandas' data alignment, where existing labels map to values, and new labels without corresponding data result in `NaN`.
*   **Advanced Filtering Capabilities**:
    *   Series elements can be filtered using multiple conditions combined with logical operators (`&` for AND, `|` for OR). For example, filtering for values greater than 5 and less than 100.
    *   The `~series.isin()` method efficiently filters out elements whose values are present in a specified list (e.g., values not in `[70, 90]`).
*   **Mathematical Operations on Series**:
    *   Standard arithmetic operations (subtraction, multiplication, division) can be applied directly to Series, performing element-wise calculations.
    *   NumPy universal functions (`np.exp()`, `np.sqrt()`, `np.abs()`) are seamlessly integrated for element-wise mathematical transformations.
*   **Value Evaluation and Metadata Retrieval**:
    *   The presence of a specific value in a Series can be checked efficiently using the `in s.values` construct (e.g., `70 in s.values` returns `True`).
    *   `idxmax()` and `idxmin()` methods provide the labels corresponding to the maximum and minimum values in the Series, respectively.
*   **Effective NaN Value Handling**:
    *   `isnull()` and `notnull()` methods accurately identify missing and non-missing values.
    *   `fillna()` allows replacing `NaN` values with a specified value (e.g., `s_nan.fillna(0)`).
    *   `dropna()` provides a straightforward way to remove Series elements containing `NaN` values.
*   **Series as Dictionary-like Objects**:
    *   Series elements can be accessed using dictionary-like key access (e.g., `myseries['red']` returning `2000.0`).
    *   The `in` operator effectively checks for the existence of a key (label) in a Series (e.g., `'blue' in myseries` returns `True`).
*   **Robust Operations Between Series with Alignment Control**:
    *   Direct arithmetic operations between two Series (e.g., `myseries - myseries2`) automatically align data by labels, resulting in `NaN` for non-matching labels. For example, `myseries - myseries2` yielded `red: 1600.0` and `yellow: -500.0`, with `NaN` for 'black', 'blue', 'green', 'orange'.
    *   Methods like `.add()` and `.sub()` with a `fill_value` parameter (e.g., `myseries.add(myseries2, fill_value=0)`) provide control over how missing labels are treated, preventing `NaN` by substituting a specified value (e.g., `0`) during the operation. This resulted in `black: 700.0` and `blue: 1000.0` in addition, and `black: -700.0` and `blue: 1000.0` in subtraction, for labels not common to both series.

### Insights or Next Steps

*   Pandas Series offers a highly flexible and intuitive interface for data manipulation, resembling both NumPy arrays and Python dictionaries, which simplifies common data tasks.
*   Understanding the data alignment behavior and the use of `fill_value` in operations between Series is crucial for robust data processing, especially when dealing with potentially misaligned or incomplete datasets.
