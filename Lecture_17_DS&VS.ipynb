#Contour Plot

A quite common type of chart in the scientific world is the contour plot or contour map. This visualization is in fact suitable for displaying three-dimensional surfaces through a contour map composed of curves closed showing the points on the surface that are located at the same level, or that have the same z value.

Although visually the contour plot is a very complex structure, its implementation is not so difficult, thanks to the matplotlib library. First, you need the function z = f (x, y) for generating a three-dimensional surface. Then, once you have defined a range of values x, y that will define the area of the map to be displayed, you can calculate the z values for each pair (x, y), applying the function f(x, y) just defined in order to obtain a matrix of z values. Finally, thanks to the contour() function you can generate the contour map of the surface. It is often desirable to add also a color map along with a contour map. That is, the areas delimited by the curves of level are filled by a color gradient, defined by a color map. For example, as in Figure, you may indicate negative values with increasingly dark shades of blue, and move to yellow and
then red with the increase of the positive values.

```python
import matplotlib.pyplot as plt
import numpy as np
dx = 0.01; dy = 0.01
x = np.arange(-2.0,2.0,dx)
y = np.arange(-2.0,2.0,dy)
X,Y = np.meshgrid(x,y)
def f(x,y):
  return (1 - y**5 + x**5)*np.exp(-x**2-y**2)
C = plt.contour(X,Y,f(X,Y),8,colors='black')
plt.contourf(X,Y,f(X,Y),8)
plt.clabel(C, inline=1, fontsize=10)
```

This code snippet uses Matplotlib and NumPy to create and display a contour plot, which is excellent for visualizing 3D surfaces on a 2D plane. Let's break it down:

*   **`import matplotlib.pyplot as plt`**: Imports the `pyplot` module from Matplotlib, commonly aliased as `plt`, for creating plots.
*   **`import numpy as np`**: Imports the `NumPy` library, aliased as `np`, which is essential for numerical operations, especially array manipulation and mathematical functions.
*   **`dx = 0.01; dy = 0.01`**: Defines the step size for x and y coordinates. These small values ensure a smooth contour.
*   **`x = np.arange(-2.0,2.0,dx)`**: Creates a NumPy array `x` ranging from -2.0 up to (but not including) 2.0, with increments of `dx` (0.01). This will be the x-axis range.
*   **`y = np.arange(-2.0,2.0,dy)`**: Similar to `x`, this creates a NumPy array `y` for the y-axis range.
*   **`X,Y = np.meshgrid(x,y)`**: This function creates 2D coordinate arrays `X` and `Y` from the 1D `x` and `y` arrays. `X` will have rows where each row is a copy of `x`, and `Y` will have columns where each column is a copy of `y`. This is crucial for evaluating a 2D function over a grid.
*   **`def f(x,y): return (1 - y**5 + x**5)*np.exp(-x**2-y**2)`**: Defines a Python function `f(x,y)` that calculates the z-value for given x and y coordinates. This function represents the 3D surface we want to visualize.
*   **`C = plt.contour(X,Y,f(X,Y),8,colors='black')`**: This is the first plotting command:
    *   `X,Y`: The 2D coordinate arrays.
    *   `f(X,Y)`: The z-values calculated by applying the function `f` to the coordinate grid.
    *   `8`: Specifies the number of contour levels to draw.
    *   `colors='black'`: Sets the color of the contour lines to black. The result is stored in `C` for later use by `plt.clabel()`.
*   **`plt.contourf(X,Y,f(X,Y),8)`**: This command creates filled contours. It uses the same `X`, `Y`, and `f(X,Y)` data and `8` levels, but instead of just drawing lines, it fills the areas between the contour lines with different colors (using Matplotlib's default colormap).
*   **`plt.clabel(C, inline=1, fontsize=10)`**: This function adds labels to the contour lines generated by `plt.contour()`:
    *   `C`: The contour object returned by `plt.contour()`.
    *   `inline=1`: Ensures that the labels are drawn directly on the contour lines, with a white background to make them readable.
    *   `fontsize=10`: Sets the font size of the labels to 10 points.

The standard color gradient (color map) is represented in Figure. Actually you choose among a large number of color maps available just specifying them with the cmap kwarg.

Furthermore, when you have to deal with this kind of representation, adding a color scale as a reference to the side of the graph is almost a must. This is possible by simply adding the function colorbar() at the
end of the code. In Figure 7-52 you can see another example of color map which starts from black, to pass through red, then yellow until reaching white for the highest values. This color map is plt.cm.hot.

```python
import matplotlib.pyplot as plt
import numpy as np
dx = 0.01; dy = 0.01
x = np.arange(-2.0,2.0,dx)
y = np.arange(-2.0,2.0,dy)
X,Y = np.meshgrid(x,y)
C = plt.contour(X,Y,f(X,Y),8,colors='black')
plt.contourf(X,Y,f(X,Y),8,cmap=plt.cm.hot)
plt.clabel(C, inline=1, fontsize=10)
plt.colorbar()
```

This code snippet builds upon the previous contour plot example by adding a specific colormap and a color bar, enhancing the visualization of the 3D surface.

Here's a breakdown of the code:

*   **`import matplotlib.pyplot as plt`**: Imports the `pyplot` module from Matplotlib for creating plots.
*   **`import numpy as np`**: Imports the `NumPy` library for numerical operations.
*   **`dx = 0.01; dy = 0.01`**: Defines step sizes for x and y coordinates, ensuring a smooth contour.
*   **`x = np.arange(-2.0, 2.0, dx)`**: Creates a NumPy array `x` for the x-axis range.
*   **`y = np.arange(-2.0, 2.0, dy)`**: Creates a NumPy array `y` for the y-axis range.
*   **`X, Y = np.meshgrid(x, y)`**: Creates 2D coordinate arrays `X` and `Y` from the 1D `x` and `y` arrays, essential for evaluating the 2D function `f(x,y)` over a grid. (The `f` function is assumed to be defined in a previous cell).
*   **`C = plt.contour(X, Y, f(X,Y), 8, colors='black')`**: Draws the contour lines in black. `C` stores the contour object, which is used later for labeling.
*   **`plt.contourf(X, Y, f(X,Y), 8, cmap=plt.cm.hot)`**: This command creates filled contours:
    *   `cmap=plt.cm.hot`: Specifies the colormap to use for filling the contours. `plt.cm.hot` is a predefined Matplotlib colormap that typically transitions from black (low values) to red, then yellow, and finally white (high values).
*   **`plt.clabel(C, inline=1, fontsize=10)`**: Adds labels directly on the contour lines, making them readable by setting them inline and with a specific font size.
*   **`plt.colorbar()`**: Adds a color bar to the side of the plot. This acts as a legend for the colormap, visually indicating the range of z-values represented by the different colors in the filled contours.

#Polar Chart

Another type of advanced chart that is having some success is the polar chart. This type of chart is characterized by a series of sectors which extend radially; each of these areas will occupy a certain angle. Thus you can display two different values assigning them to the magnitudes that characterize the polar chart: the extension of the radius r and the angle q occupied by the sector. These in fact are the polar coordinates (r, q), an alternative way of representing functions at the coordinate axes. From the graphical
point of view you could imagine it as a kind of chart that has characteristics both of the pie chart and of the bar chart. In fact as the pie chart, the angle of each sector gives percentage information represented by that category with respect to the total. As for the bar chart, the radial extension is the numerical
value of that category.

So far we have used the standard set of colors using single characters as the color code (e.g., ‘r’ to indicate red). In fact you can use any sequence of colors you want. You have to define a list of string values which contain RGB codes in the #rrggbb format corresponding to the colors you want.

Oddly, for getting a polar chart you have to use the bar() function in which you pass the list containing the angles q and a list of the radial extension of each sector. The result will be a polar chart as shown in
Figure

```python
import matplotlib.pyplot as plt
import numpy as np
N = 8
theta = np.arange(0.,2 * np.pi, 2 * np.pi / N)
radii = np.array([4,7,5,3,1,5,6,7])
plt.axes([0.025, 0.025, 0.95, 0.95], polar=True)
colors = np.array(['#4bb2c5', '#c5b47f', '#EAA228', '#579575', '#839557', '#958c12',
'#953579', '#4b5de4'])
bars = plt.bar(theta, radii, width=(2*np.pi/N), bottom=0.0, color=colors)
```

This code snippet uses Matplotlib and NumPy to create a polar bar chart. Let's break down each part:

*   **`import matplotlib.pyplot as plt`**: Imports the `pyplot` module from the `matplotlib` library, commonly aliased as `plt` for convenience. This is standard for Matplotlib plotting.
*   **`import numpy as np`**: Imports the `NumPy` library, aliased as `np`, which is essential for numerical operations, especially array creation and mathematical constants like `pi`.
*   **`N = 8`**: Defines an integer `N` as 8. This represents the number of bars (or sectors) in the polar chart.
*   **`theta = np.arange(0.,2 * np.pi, 2 * np.pi / N)`**: This line creates a `NumPy` array `theta`.
    *   `np.arange(start, stop, step)` generates a sequence of numbers. Here, it starts from 0., goes up to (but not including) `2 * np.pi` (a full circle in radians), with a step size determined by `(2 * np.pi / N)`. This effectively divides the full circle into `N` equal angles, which will be the positions of the bars.
*   **`radii = np.array([4,7,5,3,1,5,6,7])`**: This creates a `NumPy` array `radii` containing 8 numerical values. These values will determine the length (radius) of each bar in the polar chart.
*   **`plt.axes([0.025, 0.025, 0.95, 0.95], polar=True)`**: This creates a new set of axes for the plot.
    *   `[0.025, 0.025, 0.95, 0.95]` specifies the position and size of the axes within the figure as a fraction of the figure width and height (left, bottom, width, height).
    *   `polar=True` is the crucial argument that tells Matplotlib to create polar axes, meaning the plot will use radial and angular coordinates instead of Cartesian (x, y) coordinates.
*   **`colors = np.array(['#4bb2c5', '#c5b47f', '#EAA228', '#579575', '#839557', '#958c12', '#953579', '#4b5de4'])`**: This creates a `NumPy` array `colors` containing hexadecimal color codes. Each code corresponds to the color of one of the 8 bars.
*   **`bars = plt.bar(theta, radii, width=(2*np.pi/N), bottom=0.0, color=colors)`**: This is the main command to create the polar bar chart:
    *   `theta`: Provides the angular positions of the bars.
    *   `radii`: Provides the height (radial length) of each bar.
    *   `width=(2*np.pi/N)`: Sets the angular width of each bar, making them fit perfectly around the circle with no gaps (as it's the same as the step size for `theta`).
    *   `bottom=0.0`: Specifies the radial coordinate from which the bars extend. Here, they start from the center (radius 0).
    *   `color=colors`: Assigns the custom `colors` array to color each bar individually.

In this example you have defined the sequence of colors using the format #rrggbb, but you can specify a sequence of colors as strings with their actual name

```python
import matplotlib.pyplot as plt
import numpy as np
N = 8
theta = np.arange(0.,2 * np.pi, 2 * np.pi / N)
radii = np.array([4,7,5,3,1,5,6,7])
plt.axes([0.025, 0.025, 0.95, 0.95], polar=True)
colors = np.array(['lightgreen', 'darkred', 'navy', 'brown', 'violet', 'plum',
'yellow', 'darkgreen'])
bars = plt.bar(theta, radii, width=(2*np.pi/N), bottom=0.0, color=colors)
```

# 3D Surfaces

In the previous section you used the contour plot to represent the three-dimensional surfaces through the
level lines. Using the mplot3D package, surfaces can be drawn directly in 3D. In this example you will use again the same function z = f (x, y) you have used in the contour map.

Once you have calculated the meshgrid you can view the surface with the plot_surface() function. A three-dimensional surface of blue color will appear as in Figure

```python
from mpl_toolkits.mplot3d import Axes3D
import matplotlib.pyplot as plt
import numpy as np

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d') # Use add_subplot with projection for a clearer setup

X = np.arange(-2,2,0.1)
Y = np.arange(-2,2,0.1)
X,Y = np.meshgrid(X,Y)

def f(x,y):
  return (1 - y**5 + x**5)*np.exp(-x**2-y**2)

ax.plot_surface(X,Y,f(X,Y), rstride=1, cstride=1, cmap='viridis') # Added cmap

ax.set_title('3D Surface Plot of f(x,y)') # Added title
ax.set_xlabel('X-axis') # Added X-axis label
ax.set_ylabel('Y-axis') # Added Y-axis label
ax.set_zlabel('Z-axis') # Added Z-axis label

plt.show() # Added to display the plot
```

This code generates a 3D surface plot of the function $f(x,y) = (1 - y^5 + x^5) * e^{(-x^2 - y^2)}$.

**Key components and their roles:**

*   **`fig = plt.figure(figsize=(10, 8))`**: Creates a new figure with a specified size for better visibility.
*   **`ax = fig.add_subplot(111, projection='3d')`**: Adds a 3D subplot to the figure, enabling 3D plotting capabilities.
*   **`X, Y = np.meshgrid(X_1D, Y_1D)`**: Transforms 1D arrays of x and y coordinates into 2D meshgrids. This is crucial for evaluating a function over a 2D domain to get the corresponding z-values.
*   **`def f(x,y): ...`**: Defines the mathematical function that dictates the shape of the 3D surface.
*   **`ax.plot_surface(X, Y, f(X,Y), rstride=1, cstride=1, cmap='viridis')`**: Plots the 3D surface.
    *   `rstride` and `cstride` control the sampling density of the surface.
    *   `cmap='viridis'` applies a colormap (in this case, 'viridis') to the surface, visually representing the z-values with varying colors, which helps in perceiving depth and value changes.
*   **`ax.set_title(...)`, `ax.set_xlabel(...)`, `ax.set_ylabel(...)`, `ax.set_zlabel(...)`**: These functions add a descriptive title and labels to the x, y, and z axes, making the plot more understandable.
*   **`plt.show()`**: Displays the generated 3D plot.

A 3D surface stands out most by changing the color map, for example by setting the cmap kwarg. You can also rotate the surface using the view_init() function. In fact, this function adjusts the view point from
which you see the surface, changing the two kwarg elev and azim. Through their combination you can get the surface displayed from any angle. The first kwarg adjusts the height at which the surface is seen, while azim adjusts the angle of rotation of the surface.

For instance, you can change the color map using plt.cm.hot and moving the view point to elev=30 and azim=125. The result is shown in Figure.

```python
from mpl_toolkits.mplot3d import Axes3D
import matplotlib.pyplot as plt
import numpy as np # Ensure numpy is imported
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d') # Updated to modern way of adding 3D subplot
X = np.arange(-2,2,0.1)
Y = np.arange(-2,2,0.1)
X,Y = np.meshgrid(X,Y)
def f(x,y):
  return (1 - y**5 + x**5)*np.exp(-x**2-y**2)
ax.plot_surface(X,Y,f(X,Y), rstride=1, cstride=1, cmap=plt.cm.hot)
ax.view_init(elev=30,azim=125)
plt.show()
```

#Scatter Plot in 3D

However the mode most used among all 3D views remains the 3D scatter plot. With this type of visualization you can identify if the points follow particular trends, but above all if they tend to cluster.

In this case you will use the scatter() function as the 2D case but applied on Axes3D object. Doing this you can visualize different series, expressed by the calls to the scatter() function, all together in the same 3D representation

```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D
xs = np.random.randint(30,40,100)
ys = np.random.randint(20,30,100)
zs = np.random.randint(10,20,100)
xs2 = np.random.randint(50,60,100)
ys2 = np.random.randint(30,40,100)
zs2 = np.random.randint(50,70,100)
xs3 = np.random.randint(10,30,100)
ys3 = np.random.randint(40,50,100)
zs3 = np.random.randint(40,50,100)
fig = plt.figure(figsize=(10, 8)) # Added figsize for better display
ax = fig.add_subplot(111, projection='3d') # Modern way to create 3D axes
ax.scatter(xs,ys,zs)
ax.scatter(xs2,ys2,zs2,c='r',marker='^')
ax.scatter(xs3,ys3,zs3,c='g',marker='*')
ax.set_xlabel('X Label')
ax.set_ylabel('Y Label')
ax.set_zlabel('Z Label')
plt.tight_layout() # Adjust layout to prevent labels from being cut off
plt.show()
```

This code snippet generates a 3D scatter plot, which is particularly useful for visualizing relationships and clusters within three-dimensional data. Let's break down each part:

*   **`import matplotlib.pyplot as plt`**: Imports the `pyplot` module from Matplotlib for creating plots.
*   **`import numpy as np`**: Imports the `NumPy` library for numerical operations, especially for generating arrays of random numbers.
*   **`from mpl_toolkits.mplot3d import Axes3D`**: This crucial import brings in the `Axes3D` class, which is necessary to create a 3D plotting environment.

*   **`xs = np.random.randint(30,40,100)`**, `ys = np.random.randint(20,30,100)`, `zs = np.random.randint(10,20,100)`:
    These lines generate three arrays (`xs`, `ys`, `zs`), each containing 100 random integers within specified ranges. These will form the x, y, and z coordinates for the first series of data points.

*   **`xs2 = np.random.randint(50,60,100)`**, `ys2 = np.random.randint(30,40,100)`, `zs2 = np.random.randint(50,70,100)`:
    Similarly, these lines generate data for a second series of 100 random points, with different ranges to simulate a distinct cluster.

*   **`xs3 = np.random.randint(10,30,100)`**, `ys3 = np.random.randint(40,50,100)`, `zs3 = np.random.randint(40,50,100)`:
    And these lines generate data for a third series of 100 random points.

*   **`fig = plt.figure()`**: Creates a new figure object, which serves as a container for all the plot elements.

*   **`ax = Axes3D(fig)`**: Creates a 3D `Axes` object from the figure. This object provides the 3D plotting capabilities.

*   **`ax.scatter(xs,ys,zs)`**: Plots the first set of data points (`xs`, `ys`, `zs`) as a scatter plot in 3D. By default, these points will be blue circles.

*   **`ax.scatter(xs2,ys2,zs2,c='r',marker='^')`**: Plots the second set of data points. Here, `c='r'` sets the color to red, and `marker='^'` changes the marker style to upward-pointing triangles.

*   **`ax.scatter(xs3,ys3,zs3,c='g',marker='*')`**: Plots the third set of data points. `c='g'` sets the color to green, and `marker='*'` changes the marker style to asterisks.

*   **`ax.set_xlabel('X Label')`**, `ax.set_ylabel('Y Label')`, `ax.set_zlabel('Z Label')`:
    These functions set the labels for the x, y, and z axes, respectively, making the plot easier to understand.

*   **`plt.show()`**: Displays the created 3D plot. This command is essential to render the figure and show it to the user. Without it, the plot might be created in memory but not shown visually.

```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D
x = np.arange(8)
y = np.random.randint(0,10,8)
y2 = y + np.random.randint(0,3,8)
y3 = y2 + np.random.randint(0,3,8)
y4 = y3 + np.random.randint(0,3,8)
y5 = y4 + np.random.randint(0,3,8)
clr = ['#4bb2c5', '#c5b47f', '#EAA228', '#579575', '#839557', '#958c12', '#953579',
'#4b5de4']
fig = plt.figure(figsize=(10, 8)) # Set a consistent figure size
ax = fig.add_subplot(111, projection='3d') # Use add_subplot with projection for a clearer setup
ax.bar(x,y,0,zdir='y',color=clr)
ax.bar(x,y2,10,zdir='y',color=clr)
ax.bar(x,y3,20,zdir='y',color=clr)
ax.bar(x,y4,30,zdir='y',color=clr)
ax.bar(x,y5,40,zdir='y',color=clr)
ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')
ax.set_zlabel('Z Axis')
plt.tight_layout() # Adjust layout to prevent labels from being cut off
plt.show()
```

This code snippet generates a 3D bar chart, which is useful for visualizing multiple series of data in three dimensions. Let's break down each part:

*   **`import matplotlib.pyplot as plt`**, **`import numpy as np`**, **`from mpl_toolkits.mplot3d import Axes3D`**: These lines import the necessary libraries for plotting, numerical operations, and 3D visualization, respectively.
*   **`x = np.arange(8)`**: Creates a NumPy array `x` from 0 to 7. These values will serve as the x-coordinates for the bars.
*   **`y = np.random.randint(0,10,8)`** and subsequent `y` variables (`y2` to `y5`):
    These lines generate five different series of y-values. Each subsequent series builds upon the previous one by adding a random integer, simulating layered data.
*   **`clr = [...]`**: Defines a list of hexadecimal color codes. These colors will be used to distinguish the different series in the 3D bar chart.
*   **`fig = plt.figure()`**: Creates a new figure object, which acts as a container for the plot.
*   **`ax = Axes3D(fig)`**: Creates a 3D `Axes` object and adds it to the figure. This object provides the functionality for plotting in three dimensions.
*   **`ax.bar(x,y,0,zdir='y',color=clr)`** (and similar calls for `y2` through `y5`):
    These are the core commands for plotting the 3D bars:
    *   `x`: The x-coordinates for the bars.
    *   `y` (or `y2`, `y3`, etc.): The height of the bars for each series.
    *   The third argument (`0`, `10`, `20`, `30`, `40`):
        This argument, when `zdir='y'` is used, specifies the 'depth' (along the y-axis) at which each set of bars is plotted. This creates the effect of multiple layers or series extending into the depth of the chart.
    *   `zdir='y'`: This is a crucial parameter that indicates the direction in which the bars extend. Here, it means the bars are oriented along the z-axis, and their position along the y-axis is determined by the third argument.
    *   `color=clr`: Assigns a distinct color from the `clr` list to each bar series.
*   **`ax.set_xlabel('X Axis')`**, **`ax.set_ylabel('Y Axis')`**, **`ax.set_zlabel('Z Axis')`**: These functions set the labels for the respective axes, making the plot understandable.
*   **`plt.show()`**: Displays the generated 3D bar chart.

#Display Subplots within Other Subplots

Now an even more advanced method will be explained: the ability to view charts within others, enclosed within frames. Since we are talking of frames, i.e., Axes objects, you will need to separate the main Axes (i.e., the general chart) from the frame you want to add that will be another instance of Axes. To do this you use the figures() function to get the Figure object on which you will define two different Axes objects using the add_axes() function. See the result of this example in Figure

```python
import matplotlib.pyplot as plt
fig = plt.figure()
ax = fig.add_axes([0.1,0.1,0.8,0.8])
inner_ax = fig.add_axes([0.6,0.6,0.25,0.25])
```

This code snippet uses Matplotlib to create a figure with two sets of axes: a main set of axes and a smaller, inset set of axes.

1.  `import matplotlib.pyplot as plt`: This line imports the `pyplot` module from the Matplotlib library, which is a collection of functions that make Matplotlib work like MATLAB. It's conventionally aliased as `plt`.
2.  `fig = plt.figure()`: This creates a new figure. In Matplotlib, a `Figure` is the top-level container for all plot elements, including one or more `Axes` (which are the actual plotting areas).
3.  `ax = fig.add_axes([0.1,0.1,0.8,0.8])`: This line adds a set of `Axes` to the figure. The argument `[left, bottom, width, height]` specifies the position and size of the axes as a fraction of the figure's width and height. So, this `ax` will be the main plotting area, starting 10% from the left and bottom edges and taking up 80% of the figure's width and height.
4.  `inner_ax = fig.add_axes([0.6,0.6,0.25,0.25])`: This line adds another, smaller set of `Axes` to the *same* figure. This `inner_ax` is positioned at 60% from the left and bottom edges, and takes up 25% of the figure's width and height. This effectively creates an 'inset' plot, a smaller plot placed within the larger main plot defined by `ax`.

To better understand the effect of this mode of display, you can fill the previous Axes with real data as shown in Figure

```python
import matplotlib.pyplot as plt
import numpy as np
fig = plt.figure()
ax = fig.add_axes([0.1,0.1,0.8,0.8])
inner_ax = fig.add_axes([0.6,0.6,0.25,0.25])
x1 = np.arange(10)
y1 = np.array([1,2,7,1,5,2,4,2,3,1])
x2 = np.arange(10)
y2 = np.array([1,3,4,5,4,5,2,6,4,3])
ax.plot(x1,y1)
inner_ax.plot(x2,y2)
```

This code snippet uses Matplotlib to create a figure with two sets of axes: a main set of axes and a smaller, inset set of axes, and then plots data on both. Let's break down each part:

1.  **`import matplotlib.pyplot as plt`** and **`import numpy as np`**: These lines import the necessary libraries. `matplotlib.pyplot` is used for creating plots, and `numpy` is used for numerical operations, especially for creating arrays.

2.  **`fig = plt.figure()`**: This creates a new figure object. In Matplotlib, the `Figure` is the top-level container for all plot elements.

3.  **`ax = fig.add_axes([0.1,0.1,0.8,0.8])`**: This line adds the main set of `Axes` (the actual plotting area) to the figure. The list `[left, bottom, width, height]` specifies the position and size of these axes as a fraction of the figure's width and height. So, this `ax` starts 10% from the left and bottom edges and takes up 80% of the figure's width and height.

4.  **`inner_ax = fig.add_axes([0.6,0.6,0.25,0.25])`**: This adds a second, smaller set of `Axes` to the *same* figure. This `inner_ax` is positioned at 60% from the left and bottom edges and occupies 25% of the figure's width and height, creating an 'inset' plot within the main one.

5.  **`x1 = np.arange(10)`** and **`y1 = np.array([1,2,7,1,5,2,4,2,3,1])`**: These lines create data for the main plot. `x1` generates a NumPy array from 0 to 9, and `y1` is another NumPy array containing corresponding y-values.

6.  **`x2 = np.arange(10)`** and **`y2 = np.array([1,3,4,5,4,5,2,6,4,3])`**: Similar to above, these lines create a separate set of data for the inset plot.

7.  **`ax.plot(x1,y1)`**: This command plots the `x1` and `y1` data onto the main `ax` (the larger plotting area).

8.  **`inner_ax.plot(x2,y2)`**: This command plots the `x2` and `y2` data onto the `inner_ax` (the smaller, inset plotting area).

In essence, this code sets up a figure with two plots: a large one and a small one nested inside it, each displaying its own line chart. This is often used to show a zoomed-in section of the main plot or to display related but distinct data.

#Grids of Subplots

You have already seen the creation of subplots and it is quite simple using the subplots() function to add them by dividing a plot into sectors. Matplotlib allows you to manage even more complex cases using
another function called GridSpec(). This subdivision allows splitting the drawing area into a grid of sub-areas, to which you can assign one or more of them to each subplot, so that in the end you can obtain
subplots with different sizes and orientations, as you can see in Figure

```python
import matplotlib.pyplot as plt
gs = plt.GridSpec(3,3)
fig = plt.figure(figsize=(6,6))
fig.add_subplot(gs[1,:2])
fig.add_subplot(gs[0,:2])
fig.add_subplot(gs[2,0])
fig.add_subplot(gs[:2,2])
fig.add_subplot(gs[2,1:])
```

This code snippet demonstrates how to create a complex grid of subplots using `matplotlib.gridspec.GridSpec`.

1.  **`import matplotlib.pyplot as plt`**: Imports the `pyplot` module for plotting.
2.  **`gs = plt.GridSpec(3,3)`**: Creates a `GridSpec` object that defines a 3x3 grid for placing subplots.
3.  **`fig = plt.figure(figsize=(6,6))`**: Creates a new figure with a specified size.
4.  **`fig.add_subplot(gs[1,:2])`**: Adds a subplot to the figure. `gs[1,:2]` means this subplot will span across the second row (index 1) and the first two columns (indices 0 and 1) of the 3x3 grid.
5.  **`fig.add_subplot(gs[0,:2])`**: Adds another subplot, spanning the first row (index 0) and the first two columns.
6.  **`fig.add_subplot(gs[2,0])`**: Adds a subplot to the third row (index 2) and the first column (index 0).
7.  **`fig.add_subplot(gs[:2,2])`**: Adds a subplot spanning the first two rows (indices 0 and 1) and the third column (index 2).
8.  **`fig.add_subplot(gs[2,1:])`**: Adds a final subplot spanning the third row (index 2) and the second and third columns (indices 1 and 2).

This setup allows for highly flexible arrangements of subplots with varying sizes and positions within a single figure, going beyond the simple `plt.subplot()` or `plt.subplots()` functions.

Now that it’s clear to you how to manage the grid by assigning the various sectors to subplot, it’s time to see how to use these subplots. In fact, you can use the Axes object returned by each add_subplot() function to call the plot() function to draw the corresponding plot (see Figure

```python
import matplotlib.pyplot as plt
import numpy as np
gs = plt.GridSpec(3,3)
fig = plt.figure(figsize=(6,6))
x1 = np.array([1,3,2,5])
y1 = np.array([4,3,7,2])
x2 = np.arange(5)
y2 = np.array([3,2,4,6,4])
s1 = fig.add_subplot(gs[1,:2])
s1.plot(x,y,'r')
s2 = fig.add_subplot(gs[0,:2])
s2.bar(x2,y2)
s3 = fig.add_subplot(gs[2,0])
s3.barh(x2,y2,color='g')
s4 = fig.add_subplot(gs[:2,2])
s4.plot(x2,y2,'k')
s5 = fig.add_subplot(gs[2,1:])
s5.plot(x1,y1,'b^',x2,y2,'yo')
```

This code snippet demonstrates how to use a `GridSpec` layout to create a figure with multiple subplots of varying sizes and then plot different types of data on each of them.

1.  **`import matplotlib.pyplot as plt`** and **`import numpy as np`**: These lines import the necessary libraries for plotting and numerical operations.

2.  **`gs = plt.GridSpec(3,3)`**: Initializes a `GridSpec` object, defining a 3x3 grid within the figure. This grid acts as a template for arranging subplots.

3.  **`fig = plt.figure(figsize=(6,6))`**: Creates a new figure with a specified size (6x6 inches).

4.  **Data Generation**:
    *   `x1 = np.array([1,3,2,5])` and `y1 = np.array([4,3,7,2])`: Creates data for the first line plot and the last scatter plot.
    *   `x2 = np.arange(5)` and `y2 = np.array([3,2,4,6,4])`: Creates data for the bar charts and another line plot.

    *   `s1 = fig.add_subplot(gs[1,:2])`: Creates a subplot spanning the second row and the first two columns. `s1.plot(x,y,'r')` then plots data `x` and `y` (from a previous cell in the notebook) as a red line.
    *   `s2 = fig.add_subplot(gs[0,:2])`: Creates a subplot spanning the first row and the first two columns. `s2.bar(x2,y2)` plots a bar chart using `x2` and `y2` data.

This example effectively showcases how to leverage `GridSpec` for highly customized subplot layouts, allowing you to combine different visualization types within a single figure.
