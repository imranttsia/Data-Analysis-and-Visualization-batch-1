## More Arithmetic Operations Examples

```python
import numpy as np

arr1 = np.array([10, 20, 30, 40, 50])
arr2 = np.array([ 1,  2,  3,  4,  5])
scalar = 5

print(f"Array 1: {arr1}")
print(f"Array 2: {arr2}")
print(f"Scalar: {scalar}\n")

# --- Scalar Operations ---
print("--- Scalar Operations ---")

# Addition with scalar
add_scalar = arr1 + scalar
print(f"arr1 + scalar: {add_scalar}")

# Subtraction with scalar
sub_scalar = arr1 - scalar
print(f"arr1 - scalar: {sub_scalar}")

# Multiplication with scalar
mul_scalar = arr1 * scalar
print(f"arr1 * scalar: {mul_scalar}")

# Division with scalar
div_scalar = arr1 / scalar
print(f"arr1 / scalar: {div_scalar}")

# Exponentiation with scalar
pow_scalar = arr1 ** 2
print(f"arr1 ** 2: {pow_scalar}")

# Modulo with scalar
mod_scalar = arr1 % 3
print(f"arr1 % 3: {mod_scalar}\n")

# Floor Division with scalar
floor_div_scalar = arr1 // 3
print(f"arr1 // 3: {floor_div_scalar}\n")

# --- Element-wise Array Operations ---
print("--- Element-wise Array Operations ---")

# Element-wise Addition
add_arrays = arr1 + arr2
print(f"arr1 + arr2: {add_arrays}")

# Element-wise Subtraction
sub_arrays = arr1 - arr2
print(f"arr1 - arr2: {sub_arrays}")

# Element-wise Multiplication
mul_arrays = arr1 * arr2
print(f"arr1 * arr2: {mul_arrays}")

# Element-wise Division
div_arrays = arr1 / arr2
print(f"arr1 / arr2: {div_arrays}")

# Element-wise Exponentiation
pow_arrays = arr1 ** arr2
print(f"arr1 ** arr2: {pow_arrays}")

# Element-wise Modulo
mod_arrays = arr1 % arr2
print(f"arr1 % arr2: {mod_arrays}")

# Element-wise Floor Division
floor_div_arrays = arr1 // arr2
print(f"arr1 // arr2: {floor_div_arrays}\n")

# --- Comparison Operators (Element-wise) ---
print("--- Comparison Operators (Element-wise) ---")

# Greater than
gt_arrays = arr1 > arr2
print(f"arr1 > arr2: {gt_arrays}")

# Less than
lt_arrays = arr1 < arr2
print(f"arr1 < arr2: {lt_arrays}")

# Equal to
eq_arrays = arr1 == arr2
print(f"arr1 == arr2: {eq_arrays}")

# Not equal to
ne_arrays = arr1 != arr2
print(f"arr1 != arr2: {ne_arrays}")

# Greater than or equal to
ge_arrays = arr1 >= arr2
print(f"arr1 >= arr2: {ge_arrays}")

# Less than or equal to
le_arrays = arr1 <= arr2
print(f"arr1 <= arr2: {le_arrays}")
```

## NumPy Universal Functions (Ufuncs) for Arithmetic Operations

A universal function, generally called ufunc, is a function operating of an array in an element-by-element fashion. This means that it is a function that acts individually on each single element of the input array to generate a corresponding result in a new output array. At the end, you will obtain an array of the same size of the input.
There are many mathematical and trigonometric operations that meet this definition, for example, the calculation of the square root with sqrt(), the logarithm with log(), or the sin with sin().

```python
import numpy as np

arr_a = np.array([1, 2, 3, 4, 5])
arr_b = np.array([5, 4, 3, 2, 1])
scalar_val = 2

print(f"Array A: {arr_a}")
print(f"Array B: {arr_b}")
print(f"Scalar Value: {scalar_val}\n")

# --- Basic Arithmetic Ufuncs ---
print("--- Basic Arithmetic Ufuncs ---")

# Addition
print(f"np.add(arr_a, arr_b): {np.add(arr_a, arr_b)}")
print(f"np.add(arr_a, scalar_val): {np.add(arr_a, scalar_val)}\n")

# Subtraction
print(f"np.subtract(arr_a, arr_b): {np.subtract(arr_a, arr_b)}")
print(f"np.subtract(arr_a, scalar_val): {np.subtract(arr_a, scalar_val)}\n")

# Multiplication
print(f"np.multiply(arr_a, arr_b): {np.multiply(arr_a, arr_b)}")
print(f"np.multiply(arr_a, scalar_val): {np.multiply(arr_a, scalar_val)}\n")

# Division
print(f"np.divide(arr_a, arr_b): {np.divide(arr_a, arr_b)}")
print(f"np.divide(arr_a, scalar_val): {np.divide(arr_a, scalar_val)}\n")

# Modulo / Remainder
print(f"np.mod(arr_a, arr_b): {np.mod(arr_a, arr_b)}")
print(f"np.remainder(arr_a, scalar_val): {np.remainder(arr_a, scalar_val)}\n")

# Power
print(f"np.power(arr_a, arr_b): {np.power(arr_a, arr_b)}")
print(f"np.power(arr_a, scalar_val): {np.power(arr_a, scalar_val)}\n")

# Floor Division (equivalent to // operator)
print(f"np.floor_divide(arr_a, arr_b): {np.floor_divide(arr_a, arr_b)}")
print(f"np.floor_divide(arr_a, scalar_val): {np.floor_divide(arr_a, scalar_val)}\n")

# --- Other Common Mathematical Ufuncs ---
print("--- Other Common Mathematical Ufuncs ---")

# Square Root
print(f"np.sqrt(arr_a): {np.sqrt(arr_a)}\n")

# Exponential (e^x)
print(f"np.exp(arr_a): {np.exp(arr_a)}\n")

# Logarithm (natural log)
print(f"np.log(arr_a): {np.log(arr_a)}\n")

# Sine
print(f"np.sin(arr_a): {np.sin(arr_a)}\n")

# Cosine
print(f"np.cos(arr_a): {np.cos(arr_a)}\n")

# Absolute Value
neg_arr = np.array([-1, -2, 3, -4])
print(f"Array with negatives: {neg_arr}")
print(f"np.abs(neg_arr): {np.abs(neg_arr)}\n")

# Maximum (element-wise)
print(f"np.maximum(arr_a, arr_b): {np.maximum(arr_a, arr_b)}\n")

# Minimum (element-wise)
print(f"np.minimum(arr_a, arr_b): {np.minimum(arr_a, arr_b)}\n")

# Rounding
float_arr = np.array([1.2, 2.7, 3.5, 4.1])
print(f"Float Array: {float_arr}")
print(f"np.round(float_arr): {np.round(float_arr)}\n")

# Ceiling
print(f"np.ceil(float_arr): {np.ceil(float_arr)}\n")

# Floor
print(f"np.floor(float_arr): {np.floor(float_arr)}\n")
```

#Slicing

Slicing is the operation which allows you to extract portions of an array to generate new ones. Whereas using the Python lists the arrays obtained by slicing are copies, in NumPy, arrays are views onto the same underlying buffer.
Depending on the portion of the array that you want to extract (or view) you must make use of the slice syntax; that is, you will use a sequence of numbers separated by colons (‘:’) within the square brackets.

If you want to extract a portion of the array, for example one that goes from the second to the sixth element, then you have to insert the index of the starting element, that is 1, and the index of the final element, that is 5, separated by ‘:’.

```python
a = np.arange(10, 16)
a
```

```python
a[1:5]
```

Now if you want to extract from the previous portion an item, skip a specific number of following items, then extract the next, and skip again ..., you can use a third number that defines the gap in the sequence of the elements between one element and the next one to take. For example, with a value of 2, the array will take the elements in an alternating fashion.

```python
a[1:5:2]
```

To better understand the slice syntax, you also should look at cases where you do not use explicit numerical values. If you omit the first number, then implicitly NumPy interprets this number as 0 (i.e., the initial element of the array); if you omit the second number, this will be interpreted as the maximum index of the array; and if you omit the last number this will be interpreted as 1, and then all elements will be considered without intervals.

```python
print(a[::2])
print(a[:5:2])
print(a[:5:])
```

As regards the case of two-dimensional array, the slicing syntax still applies, but it is separately defined both for the rows and for the columns. For example if you want to extract only the first row:

```python
A = np.arange(10, 19).reshape((3, 3))
A
```

```python
A[0,:]
```

As you can see in the second index, if you leave only the colon without defining any number, then you will select all the columns. Instead, if you want to extract all the values of the first column, you have to write the inverse.

```python
A[:,0]
```

Instead, if you want to extract a smaller matrix then you need to explicitly define all intervals with indexes that define them.

```python
A[0:2, 0:2]
```

If the indexes of the rows or columns to be extracted are not contiguous, you can specify an array of indexes.

```python
A[[0,2], 0:2]
```
